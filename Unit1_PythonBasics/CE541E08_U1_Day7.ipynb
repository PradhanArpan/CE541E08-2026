## Section 1 — Formatted Output in Python

Professional engineering output requires aligned columns, consistent decimal places, and clear units. Python's f-strings support format specifications:

| Spec | Meaning | Example |
|---|---|---|
| `:.3f` | 3 decimal places (float) | `{3.14159:.3f}` → `3.142` |
| `:>12` | Right-align in 12-char field | `{val:>12.4f}` |
| `:<22` | Left-align in 22-char field | `{name:<22}` |
| `:^10` | Centre in 10-char field | `{text:^10}` |
| `:,.0f` | Thousands separator, no decimal | `{1250000:,.0f}` → `1,250,000` |
| `:.1%` | Percentage, 1 decimal | `{0.6734:.1%}` → `67.3%` |

## Code Block 1 — Formatting a Results Table

### What this code does

We print three engineering results (discharge, head loss, Reynolds number) in a clean, aligned table using f-string format specifications for right-alignment and decimal precision.

### Why each step is taken

**`{Q:>12.4f}`:**
Right-align the value in a 12-character field, 4 decimal places. The `>` aligns to the right edge. This makes decimal points line up vertically in a table.

**`{Re:>12,.0f}`:**
The `,` inserts thousands separators (e.g. `98,745`). `.0f` means no decimal places. This makes large numbers readable.

**`{pct:.1%}`:**
Multiplies by 100 and adds `%` sign automatically. `0.6734` becomes `67.3%`.

### Algorithm

```
1. Q=0.034521, h_f=3.14159, Re=98745.6, pct=0.6734

2. Print formatted table:
   Column 1: parameter name (left-aligned, 22 chars)
   Column 2: value (right-aligned, 12 chars, precision)
   Column 3: unit (left-aligned)
```

### Expected output

```
Parameter              Value Unit
─────────────────────────────────
Discharge Q            0.0345 m³/s
Discharge Q           34.52 L/s
Head loss               3.142 m
Reynolds number       98,746 —
```

````python
Q = 0.034521; h_f = 3.14159; Re = 98745.6

print(f"{'Parameter':<22} {'Value':>12} {'Unit'}")
print("─" * 40)
print(f"{'Discharge Q':<22} {Q:>12.4f} m³/s")
print(f"{'Discharge Q':<22} {Q*1000:>12.2f} L/s")
print(f"{'Head loss':<22} {h_f:>12.3f} m")
print(f"{'Reynolds number':<22} {Re:>12,.0f} —")
````

## Code Block 2 — input() for Interactive Calculators

### What this code does

We build an interactive Manning's pipe calculator that reads three values from the user, computes velocity and discharge, and prints a formatted result.

### Why each step is taken

**`float(input("Prompt: "))`:**
`input()` always returns a string. Wrapping with `float()` immediately converts it to a number for calculations. Always do this — never leave input as a string when arithmetic is needed.

**Why use input() in class:**
The interactive calculator can be demonstrated live: instructor enters different diameters and slopes, students see the effect on discharge immediately. This makes the Manning's formula more intuitive than a static example.

### Algorithm

```
1. diameter_mm = float(input(...))
   slope_ratio = float(input(...))
   manning_n   = float(input(...))

2. D = diameter_mm/1000
   S = 1/slope_ratio
   Compute A, P, R, V, Q

3. Print V and Q formatted
```

### Expected output (for D=300mm, 1:500, n=0.013)

```
  Velocity  : 0.633 m/s
  Discharge : 0.04473 m³/s  (44.73 L/s)
```

````python
import math
print("PIPE FLOW CALCULATOR — CE541E08")
diameter_mm = float(input("Pipe diameter (mm): "))
slope_ratio = float(input("Slope 1 in ___: "))
manning_n   = float(input("Manning's n: "))

D=diameter_mm/1000; S=1/slope_ratio
A=math.pi*(D/2)**2; P=math.pi*D; R=A/P
V=(1/manning_n)*R**(2/3)*S**0.5; Q=V*A

print()
print(f"  Velocity  : {V:.3f} m/s")
print(f"  Discharge : {Q:.5f} m³/s  ({Q*1000:.2f} L/s)")
````

## Code Block 3 — Dictionary: Monthly Rainfall Record

### What this code does

We store monthly rainfall in a dictionary (month name → value), compute annual total and anomalies (difference from normals), and print a formatted comparison table.

### Why each step is taken

**`monthly = {"Jan":4.2, "Feb":8.1, ...}`:**
A dictionary maps keys to values. Here month names (strings) map to rainfall values (floats). Dictionaries are ideal for labelled data where you need to look up values by name.

**`sum(monthly.values())`:**
`.values()` returns all the values as a view. `sum()` totals them.

**`for m in monthly`:**
Iterating over a dictionary iterates over its keys. `monthly[m]` gives the value for key `m`.

**`normals[m] - monthly[m]`:**
Looking up the same key `m` in a different dictionary — the normals record. This is the anomaly for month `m`.

### Algorithm

```
1. monthly  = {month: actual_mm, ...}
   normals  = {month: normal_mm, ...}

2. total = sum(monthly.values())
   norm_total = sum(normals.values())

3. for m in monthly:
     diff = monthly[m] - normals[m]
     print formatted row: month, actual, normal, anomaly
```

### Expected output

```
Month      Actual   Normal    Anomaly
────────────────────────────────────
Jan           4.2      5.0      -0.8
Feb           8.1      8.0      +0.1
...
Annual total: 757.1 mm (vs normal 758.0 mm)
```

````python
monthly = {
    "Jan":4.2,"Feb":8.1,"Mar":16.5,"Apr":52.3,"May":87.6,"Jun":134.2,
    "Jul":118.4,"Aug":112.7,"Sep":95.3,"Oct":71.2,"Nov":43.8,"Dec":12.6
}
normals = {
    "Jan":5.0,"Feb":8.0,"Mar":15.0,"Apr":45.0,"May":80.0,"Jun":120.0,
    "Jul":110.0,"Aug":110.0,"Sep":90.0,"Oct":75.0,"Nov":40.0,"Dec":10.0
}
total  = sum(monthly.values())
norm_t = sum(normals.values())

print(f"{'Month':<10} {'Actual':>8} {'Normal':>8} {'Anomaly':>10}")
print("─"*40)
for m in monthly:
    diff = monthly[m] - normals[m]
    sign = "+" if diff >= 0 else ""
    print(f"{m:<10} {monthly[m]:>8.1f} {normals[m]:>8.1f} {sign}{diff:>9.1f}")
print(f"
Annual total: {total:.1f} mm (vs normal {norm_t:.1f} mm)")
````

## Session Summary — Formatting and Dictionaries

| Concept | Syntax | Effect |
|---|---|---|
| Right-align | `{val:>12.3f}` | Decimal-aligned column |
| Left-align | `{name:<22}` | Name column |
| Centre | `{text:^10}` | Centred |
| Percentage | `{ratio:.1%}` | `0.67 → 67.0%` |
| Thousands | `{n:,.0f}` | `1250000 → 1,250,000` |
| User input | `float(input("prompt"))` | Interactive value |
| Dictionary | `{key: value}` | Named lookup |
| Dict values | `d.values()` | All values |
| Dict loop | `for k in d: d[k]` | Key-value access |
