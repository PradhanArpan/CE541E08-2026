## Section 1 — Variables and Data Types

A **variable** is a named storage location in memory. In Python, you do not declare a type — Python infers it from the value you assign.

| Type | What it stores | Example |
|---|---|---|
| `int` | Whole numbers | `n_pipes = 5` |
| `float` | Decimal numbers | `diameter = 0.3` |
| `str` | Text (in quotes) | `material = "concrete"` |
| `bool` | True or False | `pipe_full = True` |

**Critical**: Python is **case-sensitive**. `True` works; `true` throws a `NameError`.

---
## Code Block 1 — Creating Variables

### What this code does

We create three variables — one float, one string, one boolean — representing properties of a pipe, then print them.

### Why each step is taken

**`pipe_diameter = 0.3`:** The decimal point makes this a float. MATLAB equivalent: `pipe_diameter = 0.3;`

**`pipe_material = "concrete"`:** Text must be in quotes. Either single or double quotes work.

**`pipe_full = True`:** Boolean values must be capitalised in Python. `true` (lowercase) is a `NameError`.

**`print()`:** Displays output in the Colab cell. Without print, nothing is shown.

### Algorithm

```
1. Assign float: pipe_diameter = 0.3
2. Assign str:   pipe_material = "concrete"
3. Assign bool:  pipe_full     = True
4. print() each variable with a label
```

### Expected output

```

```

## Code Block 2 — The if Block

### What this code does

We check whether a pipe diameter meets the minimum requirement. If it does, we compute and print the cross-sectional area using `math.pi`.

### Why each step is taken

**`if pipe_diameter >= 0.1:`:** The colon is required. The indented block below runs only when the condition is True.

**`import math`:** The `math` module provides `math.pi` (3.14159...) and `math.sqrt()`. It must be imported before use — unlike MATLAB where `pi` is always available.

**`area = math.pi * (pipe_diameter/2)**2`:** The formula A = π r². `**2` is Python's power operator (MATLAB uses `^`).

**`print(f"...")`:** f-strings embed variables directly in the string. `{round(area,4)}` formats to 4 decimal places.

### Algorithm

```
1. pipe_diameter = 0.3

2. if pipe_diameter >= 0.1:     → True for 0.3
     import math
     area = math.pi * (diameter/2)^2
     print area

3. print "Check complete."      → always runs
```

### Expected output

```
Meets minimum diameter requirement.
Cross-sectional area: 0.0707 m²
Check complete.
```

````python
pipe_diameter = 0.3

if pipe_diameter >= 0.1:          # block starts here
    print("Meets minimum diameter requirement.")
    import math
    area = math.pi * (pipe_diameter/2)**2
    print(f"Cross-sectional area: {round(area,4)} m²")

print("Check complete.")           # always runs — not indented
````

## Code Block 3 — Pipe Hydraulic Properties

### What this code does

We compute the cross-sectional area, wetted perimeter, and hydraulic radius of a full circular pipe — the geometric foundation for Manning's and Darcy-Weisbach equations.

### Why each step is taken

**`radius = diameter / 2`:** Computing radius first makes the subsequent expressions cleaner and self-documenting.

**`area = math.pi * radius**2`:** A = π r²

**`perim = math.pi * diameter`:** P = π D — the wetted perimeter of a full circular pipe.

**`R = area / perim`:** Hydraulic radius R = A/P. For a full circle, R = D/4 — verified in the last print line.

**f-string formatting `{R:.6f}`:** The format spec `:.6f` gives 6 decimal places. Use this when the value is small and precision matters.

### Algorithm

```
1. diameter = 0.6 m

2. area  = π × (diameter/2)²
   perim = π × diameter
   R     = area / perim    (= D/4 for full circle)

3. Print all four properties with labels and units
4. Verify: D/4 should equal R
```

### Expected output

```
Diameter         : 600 mm
Area             : 0.282743 m²
Wetted perimeter : 1.884956 m
Hydraulic radius : 0.150000 m
D/4 check        : 0.150000 m ✓
```

````python
import math

diameter = 0.6        # 600 mm trunk main
radius   = diameter / 2
area     = math.pi * radius**2
perim    = math.pi * diameter
R        = area / perim           # hydraulic radius = D/4 for full circle

print(f"Diameter         : {diameter*1000:.0f} mm")
print(f"Area             : {area:.6f} m²")
print(f"Wetted perimeter : {perim:.6f} m")
print(f"Hydraulic radius : {R:.6f} m")
print(f"D/4 check        : {diameter/4:.6f} m ✓")
````

## Code Block 4 — Pipe Design Script

### What this code does

We compute the complete geometric properties of a 450 mm pipe — area, wetted perimeter, hydraulic radius, and hydraulic diameter — in a well-structured script with clear inputs and outputs.

### Why each step is taken

**`diameter_mm = 450`:** Using millimetres as the input is engineer-friendly (pipe catalogues use mm). The script converts internally.

**Header comment block:** The triple-line header with `═` characters is a Python convention for script documentation. It is optional but good professional practice.

### Algorithm

```
1. Input: diameter_mm = 450
2. Convert: diameter = diameter_mm / 1000
3. Compute: area, perim, R, hydraulic_diameter
4. Print formatted report
```

### Expected output

```
═══ PIPE CROSS-SECTION CALCULATOR ═══
Diameter          :  450 mm
Area              : 0.1590 m²
Wetted perimeter  : 1.4137 m
Hydraulic radius  : 0.1125 m
Hydraulic diameter: 0.4500 m
```

````python
# ═══════════════════════════════════════════════════
# Script  : Circular Pipe Cross-Section Calculator
# Author  : CE541E08 Student
# Purpose : Compute geometric properties of a pipe
# ═══════════════════════════════════════════════════
import math

# ── INPUTS ──────────────────────────────────────────
diameter_mm = 450          # pipe diameter, mm
# ────────────────────────────────────────────────────

diameter = diameter_mm / 1000
area     = math.pi * (diameter/2)**2
perim    = math.pi * diameter
R        = area / perim
D_h      = 4 * R           # hydraulic diameter (= D for full circle)

print("═"*38)
print("  PIPE CROSS-SECTION CALCULATOR")
print("═"*38)
print(f"Diameter          : {diameter_mm:>4} mm")
print(f"Area              : {area:.4f} m²")
print(f"Wetted perimeter  : {perim:.4f} m")
print(f"Hydraulic radius  : {R:.4f} m")
print(f"Hydraulic diameter: {D_h:.4f} m")
````
