## Code Block 1 — Column Alignment and Special Symbols

### What this code does

We print a station flow table with left-aligned station names, right-aligned numeric values, centred status labels, and number formatting for large values and percentages.

### Why each step is taken

**`{station:<18}`:** Left-aligns the station name in an 18-character field — all names start at the left edge.

**`{Q*1000:>12.2f}`:** Right-aligns the discharge (L/s) in a 12-char field, 2 decimal places — decimal points line up.

**`{'OK':^10}`:** Centres the status label in a 10-char field.

**`{Re:,.0f}`:** Comma as thousands separator, no decimal — makes `98745.6` → `98,746`.

**`{pct:.1%}`:** `:.1%` multiplies by 100 and adds `%` — `0.6734` → `67.3%`.

### Expected output

```
Station            Flow (L/s)  Status
──────────────────────────────────────────
Krishnarajasagara       34.52       OK
Mettur                 128.90       OK
Biligundlu               0.00      DRY

Reynolds   : 98,746
Population : 1.25e+06
Efficiency : 67.3%
```

````python
Q=0.034521; Re=98745.6; pct=0.6734; population=1250000

print(f"{'Station':<18} {'Flow (L/s)':>12} {'Status':^10}")
print("─"*42)
print(f"{'Krishnarajasagara':<18} {Q*1000:>12.2f} {'OK':^10}")
print(f"{'Mettur':<18} {128.90:>12.2f} {'OK':^10}")
print(f"{'Biligundlu':<18} {0.00:>12.2f} {'DRY':^10}")
print()
print(f"Reynolds   : {Re:,.0f}")
print(f"Population : {population:.2e}")
print(f"Efficiency : {pct:.1%}")
````

## Code Block 2 — Monsoon Anomaly Table with Trend Arrows

### What this code does

We print a 6-month monsoon comparison table with directional arrows (↑ for above-normal, ↓ for below-normal) and percentage-of-normal values.

### Why each step is taken

**`mark = "↑" if diff>0 else "↓"`:**
A ternary expression selects the arrow character. These Unicode arrows (`↑` = `↑`, `↓` = `↓`) are printable in Python f-strings.

**`{diff:>+9.1f}`:**
The `+` forces a sign character even for positive values. `+0.1` appears as `+0.1`, not `0.1`. This makes the sign immediately visible for anomaly values.

**`{pct:>7.0f}%`:**
Right-aligns the percentage in 7 chars, no decimal, with `%` appended.

### Expected output

```
══════════════════════════════════════════════════════════
  Month      Actual   Normal    Anomaly  %Normal
──────────────────────────────────────────────────────────
  Jun         134.2    120.0  ↑   +14.2     112%
  Jul         118.4    110.0  ↑    +8.4     108%
  Aug         112.7    110.0  ↑    +2.7     102%
  Sep          95.3     90.0  ↑    +5.3     106%
  Oct          71.2     75.0  ↓    -3.8      95%
  Nov          43.8     40.0  ↑    +3.8     110%
```

````python
monthly = {"Jun":134.2,"Jul":118.4,"Aug":112.7,"Sep":95.3,"Oct":71.2,"Nov":43.8}
normals = {"Jun":120.0,"Jul":110.0,"Aug":110.0,"Sep":90.0,"Oct":75.0,"Nov":40.0}

print("="*58)
print(f"  {'Month':<10} {'Actual':>8} {'Normal':>8} {'Anomaly':>10} {'%Normal':>8}")
print("─"*58)
for m in monthly:
    diff = monthly[m]-normals[m]
    pct  = monthly[m]/normals[m]*100
    mark = "↑" if diff>0 else "↓"    # up/down arrow
    print(f"  {m:<10} {monthly[m]:>8.1f} {normals[m]:>8.1f} {mark}{diff:>+9.1f} {pct:>7.0f}%")
total = sum(monthly.values())
print(f"
  Monsoon total: {total:.1f} mm")
````

## Code Block 3 — Reusable row() Helper Function

### What this code does

We define a `row()` helper function that prints one formatted line of a report table, then use it to produce a complete pipe flow design calculation report.

### Why each step is taken

**`def row(label, val, unit="")`:**
Defining a helper function eliminates repetition. Instead of typing the same f-string format seven times, we call `row()` seven times. If the format needs to change, we update one place.

**`unit=""`:**
The third parameter has a default value — if not provided, it defaults to an empty string. This is called a default argument.

**`f"  {label:<28} {val:>12}  {unit}"`:**
The label column is 28 characters wide (left-aligned). The value column is 12 characters (right-aligned). Units appear after with a leading space.

### Algorithm

```
1. def row(label, val, unit=""):
     print formatted label+value+unit

2. Compute pipe properties (D=400mm, n=0.013, S=0.0008)

3. Call row() for each output quantity:
   row("Pipe diameter", ...)
   row("Velocity", ...)
   etc.
```

### Expected output

```
════════════════════════════════════════════════════
   PIPE FLOW DESIGN CALCULATION
════════════════════════════════════════════════════
INPUTS
────────────────────────────────────────────────────
  Pipe diameter              400 mm
  Manning's n               0.013  —
  Slope                  1 in 1250  m/m
RESULTS
────────────────────────────────────────────────────
  Velocity              0.726 m/s  ✓
  Discharge             91.15 L/s
  Head loss (250m)       0.419 m
  Reynolds number       248,695  Turbulent ✓
```

````python
import math
D_mm=400; n=0.013; S=0.0008; L=650.0; f=0.018; g=9.81; nu=1.007e-6
D=D_mm/1000; A=math.pi*(D/2)**2; P=math.pi*D; R=A/P
V=(1/n)*R**(2/3)*S**0.5; Q=V*A
h_f=f*(L/D)*(V**2/(2*g)); Re=V*D/nu

def row(label, val, unit=""):
    print(f"  {label:<28} {val:>12}  {unit}")

print("═"*52)
print(f"   {'PIPE FLOW DESIGN CALCULATION':^48}")
print("═"*52)
print("INPUTS"); print("─"*52)
row("Pipe diameter",    f"{D_mm} mm")
row("Manning's n",      f"{n}",            "—")
row("Slope",            f"1 in {int(1/S)}", "m/m")
print("RESULTS"); print("─"*52)
row("Velocity",         f"{V:.3f} m/s",    "✓" if 0.6<=V<=3.0 else "✗")
row("Discharge",        f"{Q*1000:.2f} L/s")
row("Head loss (250m)", f"{h_f:.3f} m")
row("Reynolds number",  f"{Re:,.0f}",      "Turbulent ✓" if Re>4000 else "Laminar")
````

## Session Summary — Advanced Formatting

| Format spec | Effect | Example |
|---|---|---|
| `:<n` | Left-align in n chars | `{'KRS':<18}` |
| `:>n` | Right-align in n chars | `{Q:>12.4f}` |
| `:^n` | Centre in n chars | `{'OK':^10}` |
| `:+.1f` | Force sign | `{diff:>+9.1f}` |
| `:.1%` | Percentage | `{0.67:.1%}` → `67.0%` |
| `:,.0f` | Comma separator | `{98745:,.0f}` → `98,745` |
| `:.2e` | Scientific notation | `{1250000:.2e}` → `1.25e+06` |
| `def f(a, b="")` | Default argument | `b` optional |
