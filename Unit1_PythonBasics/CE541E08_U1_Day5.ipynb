## Section 1 — Python Operators

Python provides three categories of operators relevant to engineering calculations:

| Category | Operators | Examples |
|---|---|---|
| Arithmetic | `+ - * / // % **` | `9**2=81`, `347//6=57`, `347%6=5` |
| Comparison | `> >= < <= == !=` | `V >= 0.6` → True/False |
| Logical | `and or not` | `a and b` → True only if both True |

Key difference from MATLAB: Python uses `**` for powers (MATLAB uses `^`), and `//` for integer division (floor division).

## Code Block 1 — Hydraulic Jump and Integer Division

### What this code does

We compute the sequent depth for a hydraulic jump using `math.sqrt`, and use integer division `//` and modulo `%` to plan pipe ordering for a construction project.

### Why each step is taken

**`Fr1 = V1 / math.sqrt(g * y1)`:**
Froude number formula: Fr = V / √(g×y). `math.sqrt()` computes the square root. The result determines whether flow is subcritical (Fr<1) or supercritical (Fr>1).

**`y2 = (y1/2) * (math.sqrt(1 + 8*Fr1**2) - 1)`:**
The hydraulic jump sequent depth formula. `8*Fr1**2` uses Python's `**` for squaring. The parentheses control the order of operations.

**`int(total_L // section_L)`:**
`//` is floor division — it returns the integer quotient, discarding the remainder. For 347.5 ÷ 6.0, `//` gives 57 (how many full pipe sections fit).

**`total_L % section_L`:**
`%` is the modulo operator — it returns the remainder. For 347.5 ÷ 6.0, `%` gives 5.5 (the off-cut length).

### Algorithm

```
1. y1=0.4m, V1=3.2m/s, g=9.81
   Fr1 = 3.2 / sqrt(9.81 × 0.4)
   y2  = (0.4/2) × (sqrt(1+8×Fr1²) − 1)

2. total_L=347.5m, section_L=6.0m
   full      = int(347.5 // 6.0)  → 57 sections
   remainder = 347.5 % 6.0        → 5.5 m off-cut
```

### Expected output

```
Fr1 = 1.613, y2 = 0.870 m, ratio y2/y1 = 2.175
  Pipe: 57 × 6.0 m sections + 5.5 m off-cut
```

````python
import math

# Hydraulic jump — sequent depth
y1, V1, g = 0.4, 3.2, 9.81
Fr1 = V1 / math.sqrt(g * y1)
y2  = (y1/2) * (math.sqrt(1 + 8*Fr1**2) - 1)

print(f"Fr1 = {Fr1:.3f}, y2 = {y2:.3f} m, ratio y2/y1 = {y2/y1:.3f}")

# Integer division and modulo for pipe ordering
total_L   = 347.5
section_L = 6.0
full      = int(total_L // section_L)   # // = floor division (whole sections)
remainder = total_L % section_L          # % = modulo (off-cut length)
print()
print(f"  Pipe: {full} × {section_L} m sections + {remainder} m off-cut")
````

## Code Block 2 — Velocity Design Checks

### What this code does

We check whether a pipe velocity is in the acceptable design range (0.6–3.0 m/s) and whether flow is turbulent using compound boolean conditions.

### Why each step is taken

**`vel_ok = above_min and below_max`:**
Both conditions must be True for the velocity to be acceptable. Using two named intermediate booleans (`above_min`, `below_max`) makes the logic transparent.

**`turbulent = Re > 4000`:**
A comparison directly creates a boolean. Re > 4000 means fully turbulent — Manning's and Darcy-Weisbach equations are valid for turbulent flow only.

### Algorithm

```
1. V=2.8, D=0.3, nu=1.007e-6
   above_min = V >= 0.6   → True
   below_max = V <= 3.0   → True
   vel_ok    = True AND True → True

2. Re = V×D/nu
   turbulent = Re > 4000  → True
```

### Expected output

```
Velocity checks: above_min=True, below_max=True, OK=True
Reynolds: Re=834329, turbulent=True
```

````python
import math

V   = 2.8     # m/s
D   = 0.3     # m
nu  = 1.007e-6

above_min = V >= 0.6
below_max = V <= 3.0
design_ok = above_min and below_max

Re        = V * D / nu
turbulent = Re > 4000

print(f"Velocity checks: above_min={above_min}, below_max={below_max}, OK={design_ok}")
print(f"Reynolds: Re={Re:.0f}, turbulent={turbulent}")
````

## Code Block 3 — Froude Number Table

### What this code does

We compute the Froude number for four hydraulic conditions and classify each as subcritical or supercritical using a for loop with an inline conditional expression.

### Why each step is taken

**`conditions = [("name", y, V), ...]`:**
Storing related data as a list of tuples is a clean pattern for tabular data. Each tuple contains the location name, depth, and velocity for one condition.

**`cls = "Subcritical" if Fr < 1 else "Supercritical"`:**
Python's ternary (one-line if-else) expression. It produces a string value based on the condition. This is more compact than a full if-else block for simple cases.

### Algorithm

```
1. For each (name, y, V) in conditions:
   Fr  = V / sqrt(g × y)
   cls = "Subcritical" if Fr<1 else "Supercritical"
   print formatted row

2. Use f-string alignment for table formatting
```

### Expected output

```
Location              y        V       Fr Class
───────────────────────────────────────────────────────
Irrigation canal         0.8      2.1    0.750 Subcritical
Storm drain              0.3      1.8    1.050 Supercritical
Spillway toe             0.15     6.5    5.362 Supercritical
Reservoir inlet          2.0      0.4    0.090 Subcritical
```

````python
import math
g = 9.81

conditions = [
    ("Irrigation canal", 0.8, 2.1),
    ("Storm drain",      0.3, 1.8),
    ("Spillway toe",     0.15, 6.5),
    ("Reservoir inlet",  2.0, 0.4),
]

print(f"{'Location':<20} {'y':>6} {'V':>8} {'Fr':>8} {'Class'}")
print("─" * 55)
for name, y, V in conditions:
    Fr  = V / math.sqrt(g * y)
    cls = "Subcritical" if Fr < 1 else "Supercritical"
    print(f"{name:<20} {y:>6} {V:>8} {Fr:>8.3f} {cls}")
````

## Code Block 4 — Flood Warning Boolean System

### What this code does

We combine multiple boolean conditions using `and`, `or`, and `not` to produce a composite flood warning assessment — simulating a real-time alert logic used in dam safety monitoring.

### Why each step is taken

**`danger = river_level >= danger_level or (level_warn and gate_open)`:**
The danger alert is True if EITHER the river is at danger stage OR both conditions (level in warning zone AND gate is open) are true simultaneously. `or` requires only one side to be True.

**`all_clear = not level_warn and not raining`:**
`not` inverts a boolean. `all_clear` is True only when there is neither a level warning NOR heavy rain.

### Algorithm

```
1. river_level=8.4, warning=7.5, danger=9.0
   rainfall=32.0, heavy_thresh=25.0, gate_open=False

2. level_warn = 8.4 >= 7.5   → True
   raining    = 32.0 >= 25.0 → True
   danger     = 8.4 >= 9.0 OR (True AND False) → False
   all_clear  = NOT True AND NOT True → False
```

### Expected output

```
Level warning: True
Heavy rain  : True
DANGER alert: False
All clear   : False
```

````python
river_level   = 8.4
warning_level = 7.5
danger_level  = 9.0
rainfall_mmhr = 32.0
heavy_thresh  = 25.0
gate_open     = False

level_warn  = river_level >= warning_level
raining     = rainfall_mmhr >= heavy_thresh
danger      = river_level >= danger_level or (level_warn and gate_open)
all_clear   = not level_warn and not raining

print(f"Level warning: {level_warn}")
print(f"Heavy rain  : {raining}")
print(f"DANGER alert: {danger}")
print(f"All clear   : {all_clear}")
````

## Session Summary — Operators and Booleans

| Operator | Meaning | Example |
|---|---|---|
| `**` | Power | `Fr1**2` |
| `//` | Floor division (integer quotient) | `347//6 = 57` |
| `%` | Modulo (remainder) | `347%6 = 5` |
| `>=` | Greater than or equal | `V >= 0.6` |
| `==` | Equal (comparison) | `flag == "GOOD"` |
| `and` | Both must be True | `above_min and below_max` |
| `or` | At least one True | `danger or warning` |
| `not` | Inverts boolean | `not level_warn` |
| Ternary | One-line if-else | `"Sub" if Fr<1 else "Super"` |
