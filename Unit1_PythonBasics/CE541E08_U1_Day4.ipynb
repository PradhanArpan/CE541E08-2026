## Section 1 — Python's Four Core Data Types

Every value in Python has a type. Understanding types prevents bugs — you cannot add a string to a number without conversion, and logical operations require booleans.

| Type | Keyword | How to create | Example |
|---|---|---|---|
| Integer | `int` | Whole number literal | `n = 5` |
| Float | `float` | Decimal literal | `Q = 0.034` |
| String | `str` | In quotes | `station = "KRS"` |
| Boolean | `bool` | `True` or `False` | `alert = False` |

`type(x)` returns the type of any variable.

## Code Block 1 — Checking Types

### What this code does

We create four variables of different types and print their values alongside their type names.

### Why each step is taken

**`type(x).__name__`:**
`type(x)` returns a type object like `<class 'int'>`. Appending `.__name__` extracts just the string `'int'`. This is cleaner than printing the full class representation.

**Why type matters for engineering code:**
If `discharge` is accidentally stored as a string `"0.045"` (e.g. read from a CSV without conversion), then `discharge * 1000` would raise a `TypeError`. Always verify types when loading data.

### Algorithm

```
1. session_number = 4      → int
   discharge_m3s = 0.03142 → float
   station_name = "KRS..." → str
   flood_alert = False     → bool

2. print each variable with type().__name__
```

### Expected output

```
session_number  = 4  → int
discharge_m3s   = 0.03142  → float
station_name    = Krishnarajasagara  → str
flood_alert     = False  → bool
```

````python
session_number  = 4
discharge_m3s   = 0.03142
station_name    = "Krishnarajasagara"
flood_alert     = False

print(f"session_number  = {session_number}  → {type(session_number).__name__}")
print(f"discharge_m3s   = {discharge_m3s}  → {type(discharge_m3s).__name__}")
print(f"station_name    = {station_name}  → {type(station_name).__name__}")
print(f"flood_alert     = {flood_alert}  → {type(flood_alert).__name__}")
````

## Code Block 2 — Type Conversion from CSV Data

### What this code does

We simulate loading a row from a CWC CSV file — all values arrive as strings. We convert each to its correct type using `int()`, `float()`, and a boolean comparison.

### Why each step is taken

**`raw_day = "17"`:**
When Python reads a CSV file, every value is initially a string — even numbers. `"17"` is not the same as `17`. You cannot do arithmetic on a string.

**`day_num = int(raw_day)`:**
`int()` converts a string to an integer. `float()` converts to a decimal. These are the most common type conversions in data processing.

**`quality_ok = raw_flag == "TRUE"`:**
A boolean is created by comparing a string to "TRUE". The result is `True` if they match, `False` otherwise. This is how you parse a boolean from text.

### Algorithm

```
1. All values are strings: "17", "45.6", "TRUE"
2. int("17")        → 17      (integer)
   float("45.6")   → 45.6    (float)
   "TRUE" == "TRUE" → True   (boolean comparison)
3. Print before and after types
```

### Expected output

```
Raw (all strings):
  raw_day      = '17'  type: str
  raw_rainfall = '45.6'  type: str
After conversion:
  day_num     = 17   type: int
  rainfall_mm = 45.6  type: float
  quality_ok  = True  type: bool
```

````python
# Simulate what arrives when reading a CSV file
raw_day      = "17"
raw_rainfall = "45.6"
raw_flag     = "TRUE"

print("Raw (all strings):")
print(f"  raw_day      = '{raw_day}'  type: {type(raw_day).__name__}")
print(f"  raw_rainfall = '{raw_rainfall}'  type: {type(raw_rainfall).__name__}")

# Convert to correct types
day_num     = int(raw_day)
rainfall_mm = float(raw_rainfall)
quality_ok  = raw_flag == "TRUE"    # boolean from string comparison

print("After conversion:")
print(f"  day_num     = {day_num}   type: {type(day_num).__name__}")
print(f"  rainfall_mm = {rainfall_mm}  type: {type(rainfall_mm).__name__}")
print(f"  quality_ok  = {quality_ok}  type: {type(quality_ok).__name__}")
````

## Code Block 3 — String Methods

### What this code does

We clean a messy station name from a CSV (with extra spaces and mixed case), then parse a comma-separated data line into individual typed values using `.strip()`, `.title()`, and `.split()`.

### Why each step is taken

**`.strip()`:**
Removes leading and trailing whitespace. Essential when reading CSV data where field values may have extra spaces.

**`.title()`:**
Converts to title case — first letter of each word capitalised. Useful for standardising station names.

**`.replace(' ','')`:**
Removes all spaces from a string — used here to create a valid filename without spaces.

**`line.split(",")`:**
Splits a string at each comma and returns a list. The standard way to parse a CSV row without the `csv` module.

### Algorithm

```
1. station_raw = "  KRISHNARAJASAGARA  "
   .strip() → "KRISHNARAJASAGARA"
   .title() → "Krishnarajasagara"
   .replace(' ','') → used in filename

2. line = "2024-07-15,245.6,GOOD"
   .split(",") → ["2024-07-15","245.6","GOOD"]
   float(parts[1]) → 245.6
   parts[2]=="GOOD" → True
```

### Expected output

```
Raw    : '  KRISHNARAJASAGARA  '
Clean  : 'Krishnarajasagara'
File   : Cauvery_Krishnarajasagara_2024_flow.csv
  Date: 2024-07-15, Flow: 245.6 m³/s, Quality OK: True
```

````python
station_raw = "  KRISHNARAJASAGARA  "
station     = station_raw.strip().title()   # strip spaces, then title-case
river       = "Cauvery"
year        = 2024
filename    = f"{river}_{station.replace(' ','')}_{year}_flow.csv"

print(f"Raw    : '{station_raw}'")
print(f"Clean  : '{station}'")
print(f"File   : {filename}")

# Parsing a CSV data line
line  = "2024-07-15,245.6,GOOD"
parts = line.split(",")        # split on comma → list of strings
date  = parts[0]
flow  = float(parts[1])        # string "245.6" → float 245.6
flag  = parts[2] == "GOOD"     # boolean: True if flag matches
print()
print(f"  Date: {date}, Flow: {flow} m³/s, Quality OK: {flag}")
````

## Code Block 4 — Boolean Quality Control Flag

### What this code does

We compute a composite quality control flag for a rainfall reading using boolean `and` operations — combining three separate conditions into one SUSPECT flag.

### Why each step is taken

**`exceeds_max = rainfall_mm > 250.0`:**
A comparison returns a boolean directly. `exceeds_max` is either `True` or `False`.

**`big_jump = rainfall_mm > prev_day * 10`:**
A jump of more than 10× from the previous day suggests a sensor error.

**`suspect = exceeds_max and big_jump and sensor_ok`:**
`and` requires ALL conditions to be True. The flag is only raised when all three are simultaneously satisfied — reducing false alarms.

### Algorithm

```
1. rainfall_mm=287.4, prev_day=12.3, sensor_ok=True

2. exceeds_max = 287.4 > 250.0  → True
   big_jump    = 287.4 > 12.3*10 = 123  → True
   suspect     = True AND True AND True → True

3. if suspect: print warning
```

### Expected output

```
Rainfall     : 287.4 mm
Exceeds 250? : True
>10× prev?   : True
SUSPECT FLAG : True
⚠ Requires manual verification.
```

````python
rainfall_mm  = 287.4
prev_day     = 12.3
sensor_ok    = True

exceeds_max  = rainfall_mm > 250.0          # > 250 mm is unusual
big_jump     = rainfall_mm > prev_day * 10  # 10x jump from yesterday
suspect      = exceeds_max and big_jump and sensor_ok

print(f"Rainfall     : {rainfall_mm} mm")
print(f"Exceeds 250? : {exceeds_max}")
print(f">10× prev?   : {big_jump}")
print(f"SUSPECT FLAG : {suspect}")
if suspect:
    print("⚠ Requires manual verification.")
````
