## Code Block 1 — Compound Catchment and Pipe Design

### What this code does

We compute the composite peak discharge from three sub-catchments using the Rational Method, then check whether a proposed 600 mm concrete drain at slope 1:667 can carry this flow.

### Why each step is taken

**Weighted C = Σ(Ci×Ai) / Σ(Ai):**
Different sub-areas have different runoff coefficients. The weighted average accounts for the proportion each land use contributes.

**Intensity as m/s:**
The Rational Method uses SI units: Q (m³/s) = C × i (m/s) × A (m²). Converting mm/hr to m/s: divide by 1000 (mm→m) then divide by 3600 (hr→s).

**`capacity_ok = Q_drain >= Q_peak`:**
The drain must handle at least the design peak flow. Manning's gives Q_drain. If Q_drain ≥ Q_peak, the pipe is adequate.

### Algorithm

```
1. Three sub-catchments: residential, commercial, park
   Σ(Ci×Ai) / Σ(Ai) → weighted C
   intensity_ms = 48.0 / 1000 / 3600
   Q_peak = C_weighted × intensity_ms × total_m2

2. D=600mm, n=0.013, S=1:667
   A, R, V from Manning's
   Q_drain = V × A

3. capacity_ok = Q_drain >= Q_peak
   vel_ok = 0.6 <= V <= 3.0
```

### Expected output

```
═══ CATCHMENT & DRAIN DESIGN SUMMARY ═══
Catchment
  Total area   : 26.5 ha  (265000 m²)
  Weighted C   : 0.618
  Design Q     : 0.19670 m³/s  (196.70 L/s)
Drain (D=600mm)
  Velocity     : 0.866 m/s  ✓
  Capacity     : 244.68 L/s  ✓
  Adequate     : True
```

````python
# ═══════════════════════════════════════════════════════════
# CE541E08 — Unit 1 Capstone Assignment
# PROBLEM: Compound catchment → outfall pipe design
# ═══════════════════════════════════════════════════════════
import math

# ── CATCHMENT DATA ───────────────────────────────────────────
areas_ha = [8.5, 12.3, 5.7]
c_values = [0.70, 0.55, 0.80]
intensity_mmhr = 48.0

# ── PIPE DATA ────────────────────────────────────────────────
D_mm = 600; n = 0.013; S = 0.0015; L = 280.0

# ── RATIONAL METHOD ─────────────────────────────────────────
total_ha   = sum(areas_ha)
total_m2   = total_ha * 10000
C_weighted = sum(c*a for c, a in zip(c_values, areas_ha)) / total_ha
intensity_ms = intensity_mmhr / 1000 / 3600
Q_peak     = C_weighted * intensity_ms * total_m2

# ── MANNING'S PIPE DESIGN ────────────────────────────────────
D = D_mm / 1000
A = math.pi * (D/2)**2
P = math.pi * D
R = A / P
V = (1/n) * R**(2/3) * S**0.5
Q_drain = V * A

vel_ok      = 0.6 <= V <= 3.0
capacity_ok = Q_drain >= Q_peak

# ── FORMATTED OUTPUT ─────────────────────────────────────────
print("═"*42)
print("  CATCHMENT & DRAIN DESIGN SUMMARY")
print("═"*42)
print("Catchment")
print(f"  Total area   : {total_ha} ha  ({total_m2:.0f} m²)")
print(f"  Weighted C   : {C_weighted:.3f}")
print(f"  Design Q     : {Q_peak:.5f} m³/s  ({Q_peak*1000:.2f} L/s)")
print(f"Drain (D={D_mm}mm)")
print(f"  Velocity     : {V:.3f} m/s  {'✓' if vel_ok else '✗'}")
print(f"  Capacity     : {Q_drain*1000:.2f} L/s  {'✓' if capacity_ok else '✗'}")
print(f"  Adequate     : {capacity_ok}")
````
