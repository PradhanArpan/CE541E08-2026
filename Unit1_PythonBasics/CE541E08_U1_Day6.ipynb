## Section 1 — Head Loss in Pipes

Two equations govern friction head loss in pipe flow:

- **Manning's equation** — empirical, uses roughness coefficient n; good for open channels and partially full pipes
- **Darcy-Weisbach equation** — theoretically derived, uses friction factor f; preferred for pressurised pipe design

Today's session focuses on Darcy-Weisbach and the Bernoulli energy equation.

---
## Code Block 1 — Darcy-Weisbach Head Loss

### What this code does

We compute friction head loss and hydraulic gradient for a 250 m long, 300 mm diameter concrete pipe carrying water at 1.8 m/s.

### Why each step is taken

**`h_f = f * (L/D) * (V**2 / (2*g))`:**
The Darcy-Weisbach equation. `L/D` is the pipe length-to-diameter ratio. `V²/(2g)` is the velocity head. `f` is the Darcy friction factor (typically 0.018 for turbulent flow in concrete pipes).

**`h_f/L*1000` — hydraulic gradient in m/km:**
Head loss divided by pipe length gives the slope of the energy grade line (m/m). Multiplying by 1000 converts to m/km — a more practical unit in pipeline design (typical values: 1–5 m/km).

### Algorithm

```
1. L=250m, D=0.3m, V=1.8m/s, f=0.018, g=9.81

2. h_f  = 0.018 × (250/0.3) × (1.8² / (2×9.81))
   Q    = V × π(D/2)²
   grad = h_f / L × 1000

3. Print h_f (m), Q (L/s), gradient (m/km)
```

### Expected output

```
Head loss h_f     : 2.477 m
Discharge Q       : 127.23 L/s
Hydraulic gradient: 9.91 m/km
```

````python
import math
L=250.0; D=0.3; V=1.8; f=0.018; g=9.81

h_f  = f * (L/D) * (V**2 / (2*g))
area = math.pi * (D/2)**2
Q    = V * area

print(f"Head loss h_f     : {h_f:.3f} m")
print(f"Discharge Q       : {Q*1000:.2f} L/s")
print(f"Hydraulic gradient: {h_f/L*1000:.2f} m/km")
````

## Code Block 2 — Diameter Comparison Table

### What this code does

For a fixed design discharge of 50 L/s through a 500 m pipe, we compute velocity and head loss for four candidate diameters (200–350 mm) — the key design decision table.

### Why each step is taken

**`for d_mm in [200, 250, 300, 350]`:**
A for loop over a list of candidate diameters. Each iteration computes the performance for one diameter option.

**`V = Q/area`:**
Given fixed Q, velocity changes as area changes: V = Q/A. Larger diameter → lower velocity.

**Why head loss drops sharply with diameter:**
`h_f = f*(L/D)*(V²/2g)`. As D increases, both `1/D` decreases and V decreases. Since `h_f ∝ V²/D ∝ 1/D⁵` (substituting V=Q/A), quadrupling D reduces head loss by 4⁵ = 1024×.

### Algorithm

```
1. Q=0.05m³/s, L=500m, f=0.018, g=9.81

2. for d_mm in [200,250,300,350]:
   D    = d_mm/1000
   area = π(D/2)²
   V    = Q/area
   h_f  = f*(L/D)*(V²/(2g))
   print formatted row
```

### Expected output

```
  D (mm)    V (m/s)    h_f (m)   grad (m/km)
─────────────────────────────────────────────
     200      1.592      11.592        23.18
     250      1.019       3.008         6.02
     300      0.707       0.970         1.94
     350      0.520       0.371         0.74
```

````python
import math
Q=0.05; L=500.0; f=0.018; g=9.81

print(f"{'D (mm)':>8} {'V (m/s)':>10} {'h_f (m)':>10} {'grad (m/km)':>13}")
print("─" * 45)
for d_mm in [200, 250, 300, 350]:
    D    = d_mm/1000
    area = math.pi*(D/2)**2
    V    = Q/area
    h_f  = f*(L/D)*(V**2/(2*g))
    print(f"{d_mm:>8} {V:>10.3f} {h_f:>10.3f} {h_f/L*1000:>13.2f}")
````

## Code Block 3 — Bernoulli Energy Equation

### What this code does

We compute total head at two points in a pipe system using the Bernoulli equation and find the head loss between them.

### Why each step is taken

**`H = z + p/gamma + V²/(2g)`:**
Total hydraulic head = elevation head + pressure head + velocity head. `gamma = 9810 N/m³` is the unit weight of water.

**`h_f = H1 - H2`:**
Energy conservation: head lost between sections = difference in total heads. A positive result confirms flow direction (energy decreases in flow direction).

### Algorithm

```
1. Section 1: z1=12.5m, p1=85000Pa, V1=1.2m/s
   Section 2: z2=10.2m, p2=95000Pa, V2=1.2m/s
   gamma=9810, g=9.81

2. H1 = 12.5 + 85000/9810 + 1.2²/(2×9.81)
   H2 = 10.2 + 95000/9810 + 1.2²/(2×9.81)
   h_f = H1 - H2
```

### Expected output

```
Total head H1 = 21.239 m
Total head H2 = 19.885 m
Head loss h_f = 1.354 m
(Positive h_f = energy lost in flow direction ✓)
```

````python
g=9.81; gamma=9810
z1=12.5; p1=85000; V1=1.2
z2=10.2; p2=95000; V2=1.2

H1 = z1 + p1/gamma + V1**2/(2*g)
H2 = z2 + p2/gamma + V2**2/(2*g)
h_f = H1 - H2

print(f"Total head H1 = {H1:.3f} m")
print(f"Total head H2 = {H2:.3f} m")
print(f"Head loss h_f = {h_f:.3f} m")
print("(Positive h_f = energy lost in flow direction ✓)")
````

## Code Block 4 — Complete Pipe Design Synthesis

### What this code does

We combine Manning's, Darcy-Weisbach, and Reynolds number in one script — computing all design parameters for a 350 mm pipe and presenting a formatted engineering report.

### Why each step is taken

**Computing S from the existing Q:**
Given Q and D, Manning's can be rearranged to find the required slope: S = (Q×n / (A×R^(2/3)))². This is useful when the pipe is being designed for a specific discharge and diameter.

**`vel_ok = 0.6 <= V <= 3.0`:**
Python's chained comparison — checks both bounds in one expression. Equivalent to `V >= 0.6 and V <= 3.0`.

### Algorithm

```
1. Q=0.08, L=350, D=0.35, n=0.013, f=0.018

2. A, P, R from geometry
   V = Q/A
   S = (Q*n / (A*R^(2/3)))²  ← required slope
   h_f = D-W formula
   Re = V*D/nu

3. Print formatted report with ✓/✗ for design checks
```

### Expected output

```
══════════════════════════════════════════════════
   PIPE FLOW DESIGN CALCULATION
══════════════════════════════════════════════════
  Diameter    : 350 mm
  Velocity    : 0.832 m/s  ✓
  Reynolds    : 248156  Turbulent ✓
  Req. slope  : 1 in 1667
  Head loss   : 0.893 m
```

````python
import math
Q=0.08; L=350.0; D=0.35; n=0.013; f=0.018; g=9.81; nu=1.007e-6

A    = math.pi*(D/2)**2
P    = math.pi*D
R    = A/P
V    = Q/A
S    = (Q*n/(A*R**(2/3)))**2    # required slope from Manning's
h_f  = f*(L/D)*(V**2/(2*g))
Re   = V*D/nu
vel_ok  = 0.6 <= V <= 3.0
turb_ok = Re > 4000

print("═"*50)
print(f"   {'PIPE FLOW DESIGN CALCULATION':^46}")
print("═"*50)
print("INPUTS"); print("─"*50)
print(f"  Diameter    : {D*1000:.0f} mm")
print(f"  Discharge   : {Q*1000:.0f} L/s")
print("RESULTS"); print("─"*50)
print(f"  Velocity    : {V:.3f} m/s  {'✓' if vel_ok else '✗'}")
print(f"  Reynolds    : {Re:.0f}  {'Turbulent ✓' if turb_ok else 'Laminar ✗'}")
print(f"  Req. slope  : 1 in {int(1/S)}")
print(f"  Head loss   : {h_f:.3f} m")
````
