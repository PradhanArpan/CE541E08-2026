## Section 1 — Variables Are Updatable

A variable in Python is not fixed once created — you can reassign it at any time. The old value is discarded and replaced with the new one. This is how a calculation script updates intermediate results step by step.

---
## Code Block 1 — Variable Reassignment

### What this code does

We create a pipe_diameter variable, then update it to a new value, and compute the pipe area with the updated value.

### Why each step is taken

**First assignment `pipe_diameter = 0.3`:** Creates the variable and stores 0.3 in memory.

**Reassignment `pipe_diameter = 0.6`:** The old value 0.3 is gone. The variable now holds 0.6. In Excel this is like overwriting a cell.

**Computing area after update:** The area is computed using the new value. The code reads sequentially from top to bottom — earlier assignments do not retroactively affect later statements.

### Algorithm

```
1. pipe_diameter = 0.3      (stores 0.3)
   print → "0.3 m"

2. pipe_diameter = 0.6      (overwrites with 0.6)
   print → "0.6 m"

3. area = math.pi * (0.6/2)²
   print area
```

### Expected output

```
Initial: 0.3 m
Updated: 0.6 m
Area for new diameter: 0.2827 m²
```

````python
pipe_diameter = 0.3
print("Initial:", pipe_diameter, "m")

pipe_diameter = 0.6           # update — old value gone
print("Updated:", pipe_diameter, "m")

import math
area = math.pi * (pipe_diameter/2)**2
print("Area for new diameter:", round(area,4), "m²")
````

## Code Block 2 — Manning's Equation with Named Variables

### What this code does

We compute velocity and discharge for a full circular pipe using Manning's equation, with every intermediate quantity stored in a descriptively named variable.

### Why each step is taken

**Descriptive variable names:**
`pipe_diameter_m`, `manning_n`, `bed_slope` — names that tell you exactly what the variable represents and its unit. Compare to `D`, `n`, `S` which require the reader to look up the context.

**Computing all intermediates explicitly:**
Each step (`pipe_radius_m`, `flow_area_m2`, `wetted_perim_m`, `hydraulic_rad_m`) is saved in a named variable. This makes debugging easy — you can print any intermediate and verify it.

**`velocity_ms = (1/manning_n) * hydraulic_rad_m**(2/3) * bed_slope**0.5`:**
Manning's formula: V = (1/n) × R^(2/3) × S^(1/2). Python's `**` handles both integer and fractional powers.

### Algorithm

```
1. Inputs: D=0.3m, n=0.013, S=0.001

2. radius = D/2
   area   = π × radius²
   perim  = π × D
   R      = area / perim

3. V = (1/n) × R^(2/3) × S^0.5
   Q = V × area
   Q_lps = Q × 1000

4. Print V and Q (m³/s and L/s)
```

### Expected output

```
Velocity  : 0.433 m/s
Discharge : 0.03057 m³/s  (30.57 L/s)
```

````python
import math

# Well-named variables — self-documenting code
pipe_diameter_m   = 0.3
manning_n         = 0.013
bed_slope         = 0.001

pipe_radius_m     = pipe_diameter_m / 2
flow_area_m2      = math.pi * pipe_radius_m**2
wetted_perim_m    = math.pi * pipe_diameter_m
hydraulic_rad_m   = flow_area_m2 / wetted_perim_m    # = D/4

velocity_ms       = (1/manning_n) * hydraulic_rad_m**(2/3) * bed_slope**0.5
discharge_m3s     = velocity_ms * flow_area_m2
discharge_lps     = discharge_m3s * 1000

print(f"Velocity  : {velocity_ms:.3f} m/s")
print(f"Discharge : {discharge_m3s:.5f} m³/s  ({discharge_lps:.2f} L/s)")
````

## Code Block 3 — Multiple Assignment and Swap

### What this code does

We assign three variables on one line and then swap two elevation values without a temporary variable — both are Python idioms used in engineering scripts.

### Why each step is taken

**`width, depth, slope = 2.0, 0.8, 0.002`:**
Multiple assignment in one line. Python evaluates the right side first, then assigns left to right. This is equivalent to three separate assignment statements.

**`upstream, downstream = downstream, upstream`:**
Python's tuple swap. The right side `(downstream, upstream)` is evaluated first as a tuple `(103.1, 105.4)`, then unpacked into the left side. No temporary variable needed. This is cleaner than the three-line swap used in other languages.

**`+=` — augmented assignment:**
`cumulative += rain` is shorthand for `cumulative = cumulative + rain`. It is used inside loops to accumulate running totals.

### Algorithm

```
1. width, depth, slope = 2.0, 0.8, 0.002 (3 at once)

2. upstream=105.4, downstream=103.1
   upstream, downstream = downstream, upstream
   → swap: upstream=103.1, downstream=105.4

3. for rain in daily_rainfall:
     cumulative += rain
   → running total after each day
```

### Expected output

```
Width=2.0m, Depth=0.8m, Slope=0.002
Before: up=105.4, down=103.1
After:  up=103.1, down=105.4
  type(width)     = float
type('concrete')= str

Day 1   12.4           12.4
Day 2    0.0           12.4
...
Day 7   41.2           82.1
  Weekly total: 82.1 mm
```

````python
# Multiple assignment on one line
width, depth, slope = 2.0, 0.8, 0.002
print(f"Width={width}m, Depth={depth}m, Slope={slope}")

# Swap two values without a temporary variable
upstream   = 105.4
downstream = 103.1
print(f"Before: up={upstream}, down={downstream}")
upstream, downstream = downstream, upstream    # Python tuple swap
print(f"After:  up={upstream}, down={downstream}")

# type() returns the data type
print()
print(f"  type(width)     = {type(width).__name__}")
print(f"type('concrete')= {type('concrete').__name__}")
````

````python
# += augmented assignment: add and update in one step
daily_rainfall = [12.4, 0.0, 35.6, 22.1, 8.9, 0.0, 41.2]
cumulative = 0.0

print(f"{'Day':<5} {'Daily (mm)':<14} {'Cumulative (mm)'}")
print("-" * 36)
for i, rain in enumerate(daily_rainfall, 1):
    cumulative += rain          # same as: cumulative = cumulative + rain
    print(f"{i:<5} {rain:<14.1f} {cumulative:.1f}")
print()
print(f"  Weekly total: {cumulative:.1f} mm")
````

## Session Summary — Variables and Assignment

| Concept | Syntax | Notes |
|---|---|---|
| Reassignment | `x = new_value` | Old value is discarded |
| Multiple assignment | `a, b, c = 1, 2, 3` | All assigned at once |
| Swap | `a, b = b, a` | No temp variable needed |
| Augmented add | `x += v` | `x = x + v` |
| Augmented subtract | `x -= v` | `x = x - v` |
| Manning's V | `(1/n)*R**(2/3)*S**0.5` | All intermediates named |
| Q from V | `V * area` | Discharge = velocity × area |
