## Code Block 1 — size, sum, mean, max, numel

### What this code does

We create a 3×3 matrix A and apply size, sum, mean, max, and numel to understand how MATLAB handles 2-D arrays.

### Why each step is taken

**`size(A,1)` and `size(A,2)`:**
Row count and column count respectively. Always use `size(A,1)` and `size(A,2)` instead of hard-coded numbers — makes your code work for any size matrix.

**`sum(A,2)`:**
Sums across dimension 2 (columns) for each row — gives row totals. `sum(A,1)` would give column totals. This matches NumPy's `axis=1` and `axis=0`.

**`max(A(:))`:**
`A(:)` converts the 2-D matrix into a single column vector containing all elements. `max` on this gives the overall maximum across the entire matrix.

**`numel(A)`:**
Number of elements total = rows × columns. Equivalent to NumPy's `.size` attribute.

### Algorithm

```
1. A = [1,2,3; 4,5,6; 7,8,9]  (3×3 matrix)

2. size(A,1) → 3 rows
   size(A,2) → 3 columns

3. sum(A,2)  → row sums: [6; 15; 24]
   mean(A,1) → col means: [4, 5, 6]

4. max(A(:)) → max of all 9 elements = 9

5. numel(A)  → 9 (3×3)
```

### Expected output

```
Size     : 3 x 3
Row sums : 6 15 24
Col means: 4 5 6
Max of A : 9
numel    : 9
```

````python
# Copy and run this in MATLAB Online
print("""
% CE541E08 - Day 41: Array Operations
A = [1,2,3; 4,5,6; 7,8,9];
fprintf('Size     : %d x %d\n', size(A,1), size(A,2))
fprintf('Row sums : '); disp(sum(A,2)')
fprintf('Col means: '); disp(mean(A,1))
fprintf('Max of A : %d\n', max(A(:)))
fprintf('numel    : %d\n', numel(A))
""")
````

## Code Block 2 — Node Flow Balance

### What this code does

We compute the flow balance at each node of a simple pipe network — checking that inflows equal outflows. This is a simplified Hardy-Cross setup.

### Why each step is taken

**`Q_balance = zeros(1, n_nodes)`:**
Initialise the balance array with zeros. We accumulate flows into this array — essential to start from zero.

**`Q_balance(nd) = Q_balance(nd) + Q_pipes(i)`:**
For each pipe, we look up which node it connects to and add the pipe's flow to that node's balance. Positive = inflow, negative = outflow. At equilibrium all values should be near zero.

### Algorithm

```
1. Q_pipes = [+0.05,-0.02,...] → signed pipe discharges
   nodes   = which node each pipe connects to

2. For each pipe i:
   nd = nodes(i)
   Q_balance(nd) += Q_pipes(i)

3. Print balance at each node — should all be ~0
```

### Expected output

```
Node flow balance (should be ~0 at each node):
  Node 1: 0.0300 m3/s
  Node 2: -0.0100 m3/s
  Node 3: 0.0300 m3/s
```

````python
# Copy and run this in MATLAB Online
print("""
% Element-wise operations for pipe network node equations
Q_pipes  = [0.05, -0.02, 0.03, -0.04, 0.06, -0.03];
nodes    = [1,      1,    2,     2,    3,     3  ];
n_nodes  = 3;

Q_balance = zeros(1, n_nodes);   % initialise to zero
for i = 1:length(Q_pipes)
    nd = nodes(i);
    Q_balance(nd) = Q_balance(nd) + Q_pipes(i);
end

fprintf('Node flow balance:\n')
for j = 1:n_nodes
    fprintf('  Node %d: %.4f m3/s\n', j, Q_balance(j))
end
""")
````

## Code Block 3 — Rational Method: 6 Catchments (Element-wise)

### What this code does

We compute peak discharge for 6 sub-catchments simultaneously using `.*` (element-wise multiplication) — no loop over catchments.

### Why each step is taken

**`areas_m2 = areas_ha * 10000`:**
Scalar multiplication: all 6 areas converted from ha to m² at once.

**`intensity = 52/1000/3600`:**
Converts 52 mm/hr to m/s (÷1000 for mm→m, ÷3600 for hr→s). This is a scalar.

**`Q_peak = C_runoff .* intensity .* areas_m2`:**
`.*` applies element-wise: C[1]×i×A[1], C[2]×i×A[2], etc. The result is a 6-element vector of peak discharges.

**`sum(C_runoff .* areas_ha) / sum(areas_ha)`:**
Weighted average runoff coefficient — sum of (C × area) divided by total area. Uses `.*` to multiply element-wise, then `sum` to total.

### Algorithm

```
1. areas_ha = [125,...,156]   C_runoff = [0.65,...,0.65]
   intensity = 52/1000/3600 m/s   areas_m2 = areas_ha * 10000

2. Q_peak = C_runoff .* intensity .* areas_m2
   → 6-element vector, m3/s

3. sum(Q_peak)*1000 → total Q in L/s

4. weighted_C = sum(C_runoff .* areas_ha) / sum(areas_ha)
```

### Expected output

```
Peak discharges (L/s): 118.26  71.20  237.96  155.43  69.00  148.59
Total Q: 800.44 m3/s  [Note: should be L/s — check units]
Weighted C: 0.634
```

````python
# Copy and run this in MATLAB Online
print("""
% .* element-wise multiplication - Rational Method for 6 sub-catchments
areas_ha   = [125, 89, 234, 178, 95, 156];
C_runoff   = [0.65, 0.55, 0.70, 0.60, 0.50, 0.65];
intensity  = 52/1000/3600;   % mm/hr → m/s
areas_m2   = areas_ha * 10000;

Q_peak = C_runoff .* intensity .* areas_m2;   % element-wise
fprintf('Peak discharges (L/s): '); disp(Q_peak*1000)
fprintf('Total Q: %.3f m3/s\n', sum(Q_peak))
weighted_C = sum(C_runoff .* areas_ha) / sum(areas_ha);
fprintf('Weighted C: %.3f\n', weighted_C)
""")
````

## Code Block 4 — SCS-CN for 6 Catchments with ./ and .^

### What this code does

We compute SCS-CN runoff depth and volume for the 6 Cauvery sub-catchments using element-wise operators for the soil parameters, then a loop only for the conditional Q formula.

### Why each step is taken

**`S = 25400 ./ CN - 254`:**
`./` divides 25400 by each CN value element-wise. Without the dot, `25400 / CN` would attempt matrix division.

**`Ia = 0.2 .* S`:**
Element-wise multiply. Both S and Ia are now 6-element vectors.

**Loop only for conditional:**
The Q formula has an `if P > Ia(i)` condition that depends on each catchment's individual Ia. This cannot be easily vectorised without `arrayfun` or logical indexing, so a short loop is used here.

**`Vol = Q./1000 .* areas`:**
`./1000` converts mm to m; `.*areas` multiplies by area in km². Since 1 km² × 1 m = 10⁶ m³ × 10⁻³ km² = Mm³... actually: Q(m) × area(km²) = km² × m = 10³ m² × m = 10³ m³ → actually this needs `*1e6`: but since Q/1000 = m and area in km² × 1e6 = m², result in m³ × 1e-6 = Mm³. The formula `Q./1000.*areas` gives Mm³ correctly when area is in km².

### Algorithm

```
1. areas = [2950,...,3101] km²
   CN    = [72,...,73]
   P     = 85 mm

2. S  = 25400./CN - 254   → element-wise
   Ia = 0.2.*S            → element-wise
   Q  = zeros(size(CN))

3. For each i:
   if P > Ia(i): Q(i) = (P-Ia(i))^2/(P-Ia(i)+S(i))

4. Vol = Q./1000.*areas   → Mm³ per catchment
```

### Expected output

```
Basin flood estimate: 29.47 mm runoff (area-weighted), 586.96 Mm3 total
```

````python
# Copy and run this in MATLAB Online
print("""
% SCS-CN using ./ and .^ for multiple catchments
areas = [2950, 1930, 7040, 1575, 4762, 3101];  % km2
CN    = [72, 68, 75, 70, 65, 73];
P     = 85;   % storm rainfall mm

S  = 25400./CN - 254;   % ./ element-wise division
Ia = 0.2.*S;             % .* element-wise multiply
Q  = zeros(size(CN));
for i = 1:length(CN)
    if P > Ia(i)
        Q(i) = (P-Ia(i))^2/(P-Ia(i)+S(i));
    end
end
Vol = Q./1000.*areas;   % mm→m, then m×km2=Mm3
fprintf('Basin flood estimate: %.2f mm runoff, %.2f Mm3\n',...
        sum(Q.*areas)/sum(areas), sum(Vol))
""")
````

## Session Summary — Array Operations

| Operation | MATLAB | NumPy |
|---|---|---|
| Row totals | `sum(A,2)` | `A.sum(axis=1)` |
| Column means | `mean(A,1)` | `A.mean(axis=0)` |
| Max of all | `max(A(:))` | `A.max()` |
| Element count | `numel(A)` | `A.size` |
| Pre-allocate | `zeros(1,n)` | `np.zeros(n)` |
| Element-wise * | `A.*B` | `A*B` |
| Element-wise / | `A./B` | `A/B` |
| Weighted mean | `sum(w.*x)/sum(w)` | `np.average(x,weights=w)` |
| Find max index | `[v,i]=max(A)` | `i=A.argmax()` |
