### ▶ Phase II — Final Analysis

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt

# EXAMPLE COMPLETION: Topic 1 - Flood Frequency Analysis
# Replace with your actual analysis

AMS = np.array([4567,8934,3456,12340,6789,9012,5678,15670,8901,4567,
                11234,7890,5678,9012,13456,6789,4567,10234,8901,5678,
                14567,7890,6789,11234,9012,5678,8901,12345,7890,6789])
years = np.arange(1994,2024)
n = len(AMS)

# Weibull flood frequency
AMS_s = np.sort(AMS)[::-1]
rank  = np.arange(1,n+1)
P     = rank/(n+1)
T     = 1/P

print("="*58)
print("CAPSTONE — FLOOD FREQUENCY ANALYSIS")
print("Cauvery at Biligundlu (1994-2023)")
print("="*58)
print(f"n={n} years | Mean={AMS.mean():.0f} | Std={AMS.std():.0f} m3/s")
print()
print(f"{'T-year':>8} {'Q (m3/s)':>12} {'Interpretation'}")
print("-"*50)
for T_t in [2,5,10,25,50]:
    Q_T = np.interp(1/T_t,P[::-1],AMS_s[::-1])
    print(f"{T_t:>8}  {Q_T:>12.0f}  Design flood for {T_t}-year ARI")

# Plot
fig, axes = plt.subplots(1,2,figsize=(14,5))
axes[0].bar(years,AMS,color='steelblue',alpha=0.8)
axes[0].axhline(AMS.mean(),color='red',linestyle='--',label='Mean')
axes[0].set_title('Annual Maximum Series'); axes[0].legend()
axes[1].semilogy(P*100,AMS_s,'bo-',markersize=4)
for T_t,col in [(10,'orange'),(25,'red'),(50,'darkred')]:
    Qt=np.interp(1/T_t,P[::-1],AMS_s[::-1])
    axes[1].axhline(Qt,color=col,linestyle='--',label=f'T={T_t}yr: {Qt:.0f}')
axes[1].set_xlabel('Exceedance Probability (%)'); axes[1].set_ylabel('Q (m3/s)')
axes[1].set_title('Flood Duration Curve'); axes[1].legend()
plt.tight_layout(); plt.show()
```

### ▶ Design Recommendation

```python
# Every capstone MUST include an engineering recommendation
# This connects the analysis to actual civil engineering practice

print("="*58)
print("ENGINEERING RECOMMENDATION")
print("="*58)
print()
print("Project   : Spillway design for proposed dam on Cauvery")
print("Location  : Downstream of Biligundlu gauge")
print()
print("Design criteria:")
print("  Spillway design flood (SDF) = PMF or T=1000-yr")
print("  Normal design flood         = T=100-yr")
print("  Warning level trigger       = T=10-yr")
print()
# From our analysis
AMS = np.array([4567,8934,3456,12340,6789,9012,5678,15670,8901,4567,
                11234,7890,5678,9012,13456,6789,4567,10234,8901,5678,
                14567,7890,6789,11234,9012,5678,8901,12345,7890,6789])
AMS_s=np.sort(AMS)[::-1]; n=len(AMS); P=np.arange(1,n+1)/(n+1)
Q10  = np.interp(0.1, P[::-1], AMS_s[::-1])
Q25  = np.interp(0.04,P[::-1], AMS_s[::-1])
Q100 = np.interp(0.01,P[::-1], AMS_s[::-1])
print(f"Estimated T=10  yr flood : {Q10:.0f} m3/s")
print(f"Estimated T=25  yr flood : {Q25:.0f} m3/s")
print(f"Estimated T=100 yr flood : {Q100:.0f} m3/s")
print()
print("Recommendation: Design spillway for Q_100 = {:.0f} m3/s".format(Q100))
print("                with 30% freeboard factor")
print("                = {:.0f} m3/s design capacity".format(Q100*1.3))
print()
print("Limitation: 30-year record is below recommended 50+ years")
print("            Extrapolation beyond T=25 yr has high uncertainty")
print("            Recommend frequency analysis review when record extends to 50+ years")
```
