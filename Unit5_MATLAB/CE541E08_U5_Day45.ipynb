```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ── EXAMPLE COMPLETION: Topic 1 - Flood Frequency Analysis ─────────────────
# Replace with your actual analysis

AMS = np.array([4567,8934,3456,12340,6789,9012,5678,15670,8901,4567,
                11234,7890,5678,9012,13456,6789,4567,10234,8901,5678,
                14567,7890,6789,11234,9012,5678,8901,12345,7890,6789])
years = np.arange(1994, 2024)
n     = len(AMS)

# Weibull flood frequency
AMS_s = np.sort(AMS)[::-1]
rank  = np.arange(1, n+1)
P     = rank/(n+1)
T     = 1/P

print("=" * 58)
print("CAPSTONE — FLOOD FREQUENCY ANALYSIS")
print("Cauvery at Biligundlu — 30-Year AMS (1994-2023)")
print("=" * 58)
print(f"n = {n} years   Mean = {AMS.mean():.0f} m3/s   Std = {AMS.std():.0f} m3/s")
print(f"Min = {AMS.min():.0f}   Max = {AMS.max():.0f}   CV = {AMS.std()/AMS.mean():.3f}")
print()

# Design floods by interpolation
print("Weibull Design Floods:")
for T_target in [2, 5, 10, 25, 50]:
    Q_T = np.interp(1/T_target, P[::-1], AMS_s[::-1])
    print(f"  T={T_target:>3}-yr: {Q_T:.0f} m3/s")
```

```python
import numpy as np, matplotlib.pyplot as plt

AMS = np.array([4567,8934,3456,12340,6789,9012,5678,15670,8901,4567,
                11234,7890,5678,9012,13456,6789,4567,10234,8901,5678,
                14567,7890,6789,11234,9012,5678,8901,12345,7890,6789])
n     = len(AMS)
AMS_s = np.sort(AMS)[::-1]
P     = np.arange(1,n+1)/(n+1)
T     = 1/P

fig, axes = plt.subplots(1, 2, figsize=(14,5))

# Flood frequency curve
axes[0].semilogx(T, AMS_s, 'bo-', markersize=5, label='Observed')
for T_t in [10, 25, 50]:
    Q_t = np.interp(1/T_t, P[::-1], AMS_s[::-1])
    axes[0].axvline(T_t, color='grey', linestyle=':', alpha=0.5)
    axes[0].annotate(f'T={T_t}yr
{Q_t:.0f}', xy=(T_t, Q_t),
                     xytext=(T_t*1.3, Q_t*0.9), fontsize=8)
axes[0].set_xlabel('Return Period T (years) — log scale')
axes[0].set_ylabel('Peak Discharge (m3/s)')
axes[0].set_title('Flood Frequency Curve — Cauvery at Biligundlu')
axes[0].grid(which='both', alpha=0.3); axes[0].legend()

# Annual time series with trend
years = np.arange(1994, 2024)
z = np.polyfit(years, AMS, 1)
p = np.poly1d(z)
axes[1].bar(years, AMS, color='steelblue', alpha=0.7)
axes[1].plot(years, p(years), 'r-', linewidth=2, label=f'Trend {z[0]:+.0f} m3/s/yr')
axes[1].axhline(AMS.mean(), color='orange', linestyle='--', label=f'Mean {AMS.mean():.0f}')
axes[1].set_xlabel('Year'); axes[1].set_ylabel('Peak Discharge (m3/s)')
axes[1].set_title('Annual Maximum Series with Trend')
axes[1].legend(); plt.tight_layout(); plt.show()
```
