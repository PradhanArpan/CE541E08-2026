### Cell 2 - Manning's equation in Command Window

```python
# MATLAB code - run this in MATLAB Online at matlab.mathworks.com
print("""
% Manning's equation - pipe flow
D = 0.3; n = 0.013; S = 0.001;
A = pi * (D/2)^2;
P = pi * D;
R = A / P;
V = (1/n) * R^(2/3) * S^(1/2);
Q = V * A;
fprintf('Velocity V : %.3f m/s\n', V)
fprintf('Discharge Q: %.4f m3/s = %.2f L/s\n', Q, Q*1000)
""")
```

### Cell 3 - Your first .m script

```python
# MATLAB code - run this in MATLAB Online at matlab.mathworks.com
print("""
% Day37_Manning.m - save this as a script and run with F5
clc; clear;
D_mm = 300; manning_n = 0.013; bed_slope = 0.001;
D = D_mm / 1000;
A = pi * (D/2)^2; P = pi * D; R = A / P;
V = (1/manning_n) * R^(2/3) * bed_slope^(1/2);
Q = V * A;
fprintf('=== PIPE FLOW CALCULATION ===\n')
fprintf('Diameter : %d mm\n', D_mm)
fprintf('Velocity : %.3f m/s\n', V)
fprintf('Discharge: %.4f m3/s = %.2f L/s\n', Q, Q*1000)
""")
```

### Cell 4 - Darcy-Weisbach head loss

```python
# MATLAB code - run this in MATLAB Online at matlab.mathworks.com
print("""
% Darcy-Weisbach head loss
L = 250; D = 0.3; V = 1.8; f = 0.018; g = 9.81;
h_f = f * (L/D) * (V^2 / (2*g));
grad = h_f / L * 1000;
fprintf('Head loss h_f : %.3f m\n', h_f)
fprintf('Hyd gradient  : %.2f m/km\n', grad)
""")
```
