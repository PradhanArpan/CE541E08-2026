## Section 1 — Why MATLAB?

Python (Units 1–4) is a general-purpose programming language. MATLAB is a specialised environment designed for numerical computing, matrix operations, and engineering analysis. In civil engineering practice you will encounter both:

| Feature | Python | MATLAB |
|---|---|---|
| General programming | ✓ | ✓ |
| Matrix operations | NumPy | Built-in (everything is a matrix) |
| Data analysis | Pandas | Tables, Statistics Toolbox |
| Cost | Free | Licensed (free for students via MathWorks) |
| Industry use | Data science, automation | Structural, hydraulic, geotechnical analysis |

**Key MATLAB facts:**
- Every value in MATLAB is a matrix — a scalar is a 1×1 matrix, a vector is a 1×n matrix
- MATLAB uses 1-based indexing (first element is index 1, not 0)
- Lines ending with `;` suppress output; without `;`, MATLAB prints the result
- `%` starts a comment (Python uses `#`)
- Scripts are saved as `.m` files and run with F5 or the Run button

## Code Block 1 — First Commands in the Command Window

### What this code does

We type basic arithmetic and variable assignments directly in the MATLAB Command Window to get familiar with the environment. This is the equivalent of Day 1 in Python.

### Why each step is taken

**`3 + 4` (no semicolon):**
MATLAB immediately evaluates and prints the result: `ans = 7`. `ans` is a special variable that always holds the last unassigned result.

**`pipe_diameter = 0.3;` (with semicolon):**
Creates a variable and stores 0.3. The semicolon suppresses output — the value is stored but not printed. Without the semicolon, MATLAB would print `pipe_diameter = 0.3000`.

**`whos`:**
Lists all variables in the Workspace: name, size, bytes, and class. This is the MATLAB equivalent of Python's `type()` check — use it after creating variables to confirm they are what you expect.

### Algorithm

```
1. Type arithmetic: 3+4, 10*2.5, 9^2, sqrt(16)
   MATLAB prints each result as ans = ...

2. Assign variables:
   pipe_diameter = 0.3;   (semicolon suppresses output)
   manning_n     = 0.013;

3. whos → lists all variables in the Workspace
```

### Expected output (in MATLAB Command Window)

```
ans = 7
ans = 25
ans = 81
ans = 4
  Name            Size    Bytes  Class
  manning_n       1x1         8  double
  pipe_diameter   1x1         8  double
```

## Code Block 2 — Manning's Equation in the Command Window

### What this code does

We compute Manning's discharge for a full circular pipe step by step in the Command Window. Each intermediate result is printed (no semicolons) so you can verify the calculation at each step.

### Why each step is taken

**`pi`:**
MATLAB has `pi` as a built-in constant (3.14159...) — no import needed, unlike Python where you need `import math` or `import numpy as np`.

**`^` for powers:**
MATLAB uses `^` for scalar exponentiation. For arrays, `.^` is used (element-wise). This is different from Python where `**` is used for both.

**`fprintf('...\n', value)`:**
MATLAB's formatted print function. `%.3f` means 3 decimal places. `\n` is a newline. The syntax is similar to C's printf — different from Python's f-strings but the formatting codes are the same.

### Algorithm

```
1. D = 0.3 m,  n = 0.013,  S = 0.001

2. A = pi * (D/2)^2    → cross-sectional area, m²
   P = pi * D           → wetted perimeter, m
   R = A / P            → hydraulic radius, m (= D/4 for full pipe)

3. V = (1/n) * R^(2/3) * S^(1/2)   → Manning's velocity, m/s
   Q = V * A                          → discharge, m³/s

4. fprintf to print V and Q
```

### Expected output

```
Velocity V : 0.433 m/s
Discharge Q: 0.0306 m3/s = 30.58 L/s
```

````python
# Copy and run this in MATLAB Online
print("""
% Manning's equation - pipe flow
D = 0.3; n = 0.013; S = 0.001;

A = pi * (D/2)^2;   % cross-sectional area, m^2
P = pi * D;          % wetted perimeter, m
R = A / P;           % hydraulic radius, m  (= D/4 for full pipe)
V = (1/n) * R^(2/3) * S^(1/2);   % Manning's velocity, m/s
Q = V * A;           % discharge, m^3/s

fprintf('Velocity V : %.3f m/s\n', V)
fprintf('Discharge Q: %.4f m3/s = %.2f L/s\n', Q, Q*1000)
""")
````

## Code Block 3 — Manning's Script File (Day37_Manning.m)

### What this code does

We write our first MATLAB **script** — a saved `.m` file that runs all commands in sequence. This is equivalent to a Python `.py` file.

### Why each step is taken

**`clc; clear;`:**
Standard first line for any MATLAB script:
- `clc` — clears the Command Window (fresh output each run)
- `clear` — removes all variables from the Workspace (prevents leftover values from a previous run contaminating the current one)

**`D_mm = 300;`:**
Using millimetres as input is more natural for pipe design. The conversion to metres `D = D_mm/1000` happens inside the script — this is explicit and self-documenting.

**`fprintf('=== PIPE FLOW CALCULATION ===\n')`:**
Print a header line to make the output readable. Good practice in engineering scripts so the user knows which calculation is being displayed.

### Algorithm

```
1. clc; clear;   → fresh start

2. Input parameters: D_mm, manning_n, bed_slope

3. Convert: D = D_mm/1000

4. Compute: A, P, R, V, Q (same as Code Block 2)

5. fprintf formatted output:
   header line
   diameter in mm
   velocity in m/s
   discharge in m³/s and L/s
```

### Expected output

```
=== PIPE FLOW CALCULATION ===
Diameter : 300 mm
Velocity : 0.433 m/s
Discharge: 0.0306 m3/s = 30.58 L/s
```

````python
# Copy and run this in MATLAB Online
print("""
% Day37_Manning.m - save as a script and run with F5
clc; clear;

D_mm      = 300;     % pipe diameter, mm
manning_n = 0.013;   % roughness coefficient (concrete)
bed_slope = 0.001;   % longitudinal slope

D = D_mm / 1000;     % convert mm to m

A = pi * (D/2)^2;    % cross-sectional area, m^2
P = pi * D;           % wetted perimeter, m
R = A / P;            % hydraulic radius, m
V = (1/manning_n) * R^(2/3) * bed_slope^(1/2);   % velocity, m/s
Q = V * A;            % discharge, m^3/s

fprintf('=== PIPE FLOW CALCULATION ===\n')
fprintf('Diameter : %d mm\n', D_mm)
fprintf('Velocity : %.3f m/s\n', V)
fprintf('Discharge: %.4f m3/s = %.2f L/s\n', Q, Q*1000)
""")
````

## Code Block 4 — Darcy-Weisbach Head Loss

### What this code does

We compute the Darcy-Weisbach friction head loss for a pipe and the hydraulic gradient — the rate of head loss per unit length.

### Why each step is taken

**`h_f = f * (L/D) * (V^2 / (2*g))`:**
The Darcy-Weisbach equation. All variables on one line — MATLAB evaluates right-to-left following operator precedence. `^2` applies to V only (not the whole expression), so parentheses are needed: `V^2`.

**`grad = h_f / L * 1000`:**
Hydraulic gradient in m/km. Dividing head loss by pipe length gives m/m; multiplying by 1000 gives m/km — a more practical unit for pipe design.

### Algorithm

```
1. L=250m, D=0.3m, V=1.8m/s, f=0.018, g=9.81

2. h_f = f * (L/D) * (V^2 / (2*g))
   grad = h_f / L * 1000   (m/km)

3. Print h_f and grad
```

### Expected output

```
Head loss h_f : 2.477 m
Hyd gradient  : 9.91 m/km
```

````python
# Copy and run this in MATLAB Online
print("""
% Darcy-Weisbach head loss
L = 250; D = 0.3; V = 1.8; f = 0.018; g = 9.81;

h_f  = f * (L/D) * (V^2 / (2*g));
grad = h_f / L * 1000;   % m/km

fprintf('Head loss h_f : %.3f m\n', h_f)
fprintf('Hyd gradient  : %.2f m/km\n', grad)
""")
````

## Day 37 Assignment

Write a MATLAB script `Day37_Assignment.m` that:

1. Accepts diameter (mm), slope ratio (e.g. 500 for 1:500), and Manning's n from the user using `input()`
2. Computes V and Q using Manning's equation
3. Prints a formatted output
4. Checks if the velocity is in the acceptable range (0.6–3.0 m/s) using if-else

````python
# Copy and run this in MATLAB Online
print("""
% Day37_Assignment.m
clc; clear;
D_mm        = input('Enter pipe diameter (mm): ');
slope_ratio = input('Enter slope ratio (e.g. 500 for 1:500): ');
n           = input('Enter Manning n (e.g. 0.013): ');

D = D_mm/1000; S = 1/slope_ratio;
A = pi*(D/2)^2; R = D/4;
V = (1/n)*R^(2/3)*S^0.5;
Q = V*A;
fprintf('Velocity  : %.3f m/s\n', V)
fprintf('Discharge : %.2f L/s\n', Q*1000)

if V >= 0.6 && V <= 3.0
    fprintf('Design: ACCEPTABLE\n')
else
    fprintf('Design: REVIEW REQUIRED\n')
end
""")
````
