## Section 1 — MATLAB Data Types and Array Structure

In Python, you must import NumPy to work with arrays. In MATLAB, **everything is an array by default** — a scalar is a 1×1 matrix, a row vector is a 1×n matrix, a column vector is n×1, and a 2-D matrix is m×n.

MATLAB has four commonly used data types:

| Type | MATLAB | Example |
|---|---|---|
| Decimal number | `double` | `discharge = 0.045` |
| Text | `char` | `pipe_mat = 'concrete'` |
| True/False | `logical` | `data_ok = true` |
| Integer | `int32`, `uint8` | `n_pipes = int32(6)` |

Unlike Python, MATLAB does not require you to specify the type — it infers from the value.

### Why each step is taken

## Code Block 2 — Vectors and the Colon Operator

### What this code does

We create a row vector of 12 monthly rainfall values and use MATLAB's built-in functions (sum, max, min) to analyse it. We also create sequences using the colon operator.

### Why each step is taken

**`monthly_rain = [8,12,...,13]`:**
Square brackets create a vector. Commas or spaces separate elements. MATLAB creates a **row vector** by default (1×12). Semicolons inside brackets create rows of a matrix.

**`monthly_rain(6:9)`:**
MATLAB uses `()` for indexing (not `[]` like Python/NumPy). `6:9` is the colon operator producing indices 6, 7, 8, 9 (1-based). This selects June through September — the monsoon season.

**`[peak_val, peak_idx] = max(monthly_rain)`:**
MATLAB's `max` can return two outputs: the maximum value and its index. Square brackets on the left side capture multiple return values — this is unique to MATLAB syntax.

**`1:30` and `0.001:0.001:0.01`:**
The colon operator `start:step:stop` generates a sequence. `1:30` gives [1,2,...,30]. `0.001:0.001:0.01` gives [0.001, 0.002, ..., 0.010]. This is MATLAB's equivalent of `np.arange`.

### Algorithm

```
1. monthly_rain = [8,12,...,13]   → 1×12 row vector

2. sum(monthly_rain)              → annual total
   sum(monthly_rain(6:9))        → monsoon total (index 6-9)
   [peak_val, peak_idx] = max()  → peak value and its month index

3. days   = 1:30       → [1,2,...,30]
   slopes = 0.001:0.001:0.01 → 10 slope values
   length(days) → 30
```

### Expected output

```
Annual total : 765 mm
Monsoon (6-9): 460 mm
Max month    : 134 mm (month 6)
Days: 30 values, Slopes: 10 values
```

````python
# Copy and run this in MATLAB Online
print("""
% Vectors in MATLAB
monthly_rain = [8,12,18,52,87,134,118,113,95,71,44,13];
fprintf('Annual total : %.0f mm\n', sum(monthly_rain))
fprintf('Monsoon (6-9): %.0f mm\n', sum(monthly_rain(6:9)))

% [val, idx] = max() returns BOTH the value and its position
[peak_val, peak_idx] = max(monthly_rain);
fprintf('Max month    : %.0f mm (month %d)\n', peak_val, peak_idx)

% Colon operator: start:step:stop (or start:stop for step=1)
days   = 1:30;              % [1,2,...,30]
slopes = 0.001:0.001:0.01;  % [0.001,0.002,...,0.010]
fprintf('Days: %d values, Slopes: %d values\n', length(days), length(slopes))
""")
````

## Code Block 3 — 2-D Matrices

### What this code does

We create a 3×12 rainfall matrix (3 years × 12 months) and use `size()`, `sum()`, and `mean()` with dimension arguments to compute annual totals and monthly means — the MATLAB equivalent of NumPy's `axis` parameter.

### Why each step is taken

**Semicolons inside `[]` to separate rows:**
In MATLAB, a semicolon inside square brackets ends one row and starts the next. `[1,2,3; 4,5,6]` creates a 2×3 matrix. This is different from nested lists in Python.

**`size(rain_data,1)` and `size(rain_data,2)`:**
`size(A,1)` = number of rows; `size(A,2)` = number of columns. In Python/NumPy this is `data.shape[0]` and `data.shape[1]`.

**`sum(rain_data,2)`:**
In MATLAB, the second argument to `sum` specifies the dimension: `1` = sum down columns (across rows), `2` = sum across rows (across columns). `sum(rain_data,2)` gives one annual total per row. This is the MATLAB equivalent of `data.sum(axis=1)`.

**`rain_data(2,6)`:**
Access element at row 2, column 6 — Year 2, June. MATLAB uses **1-based indexing**: row 2 is the second year (2023), column 6 is June.

### Algorithm

```
1. rain_data = 3×12 matrix (semicolons separate rows)

2. size(rain_data,1) → number of rows (3)
   size(rain_data,2) → number of columns (12)

3. sum(rain_data,2)  → sum across columns → annual total per year
   mean(rain_data,1) → mean across rows   → climatological mean per month

4. rain_data(2,6)    → row 2, col 6 = Year2, June
```

### Expected output

```
Size: 3 x 12
Annual totals:  765   756   783
Monthly means:  7.7  11.7  18.7  51.7  87.3 135.7  ...
June 2023 (row2,col6): 92 mm
```

````python
# Copy and run this in MATLAB Online
print("""
% Matrices (2-D arrays) - 3 years x 12 months
rain_data = [
     8, 12,18,52,87,134,118,113,95,71,44,13;   % Year 1
     5,  8,22,48,92,145,123,108,89,68,38,10;   % Year 2
    10, 15,16,55,83,128,115,119,98,74,48,16;   % Year 3
];

fprintf('Size: %d x %d\n', size(rain_data,1), size(rain_data,2))

% sum(mat,2): sum across dim 2 (columns) → one total per row
fprintf('Annual totals: '); disp(sum(rain_data,2)')

% mean(mat,1): mean across dim 1 (rows) → one mean per column
fprintf('Monthly means: '); disp(mean(rain_data,1))

% Element access: (row, col) — 1-based indexing
fprintf('June 2023 (row2,col6): %.0f mm\n', rain_data(2,6))
""")
````

## Code Block 4 — Pipe Network as Matrix

### What this code does

We store a 6-pipe network as a matrix (each row = one pipe, columns = Length, Diameter, n, Slope) and use a for loop to compute the velocity for each pipe.

### Why each step is taken

**Matrix as data table:**
Storing network data as a matrix is a natural MATLAB pattern. Each row is one pipe; each column is one attribute. `pipes(i,2)` gives the diameter of pipe i.

**`size(pipes,1)`:**
Returns the number of rows — the number of pipes. Using this instead of a hard-coded `6` means the loop automatically handles any number of pipes.

**`for i=1:size(pipes,1)`:**
MATLAB for loop syntax: `for variable = start:stop`. The loop variable `i` takes values 1, 2, 3, 4, 5, 6. No `in` keyword, no `range()`.

### Algorithm

```
1. pipes = 6×4 matrix
   [Length, Diameter_mm, n, Slope] per row

2. for i = 1:size(pipes,1)   % i = 1 to 6
   D_mm = pipes(i,2)
   n    = pipes(i,3)
   S    = pipes(i,4)
   D    = D_mm/1000; R = D/4
   V    = (1/n)*R^(2/3)*S^0.5
   fprintf(...)
end
```

### Expected output

```
Pipe  D(mm)  V(m/s)
   1    300   0.433
   2    250   0.642
   3    200   0.792
   4    300   0.433
   5    200   0.669
   6    250   0.525
```

````python
# Copy and run this in MATLAB Online
print("""
% Pipe network as matrix
% columns: [Length_m, Diameter_mm, Manning_n, Slope]
pipes = [200,300,0.013,0.001; 150,250,0.013,0.002; 180,200,0.015,0.003;
         120,300,0.013,0.001;  90,200,0.015,0.002; 100,250,0.013,0.0015];

fprintf('%4s %6s %8s\n','Pipe','D(mm)','V(m/s)')
for i = 1:size(pipes,1)
    D_mm = pipes(i,2); n = pipes(i,3); S = pipes(i,4);
    D = D_mm/1000; R = D/4;
    V = (1/n)*R^(2/3)*S^0.5;
    fprintf('%4d %6d %8.3f\n', i, D_mm, V)
end
""")
````

## Session Summary — MATLAB Data Types and Arrays

| Concept | MATLAB | Python/NumPy |
|---|---|---|
| Scalar | `x = 5.0` (double by default) | `x = 5.0` |
| Row vector | `v = [1,2,3,4]` | `np.array([1,2,3,4])` |
| Column vector | `v = [1;2;3;4]` | `np.array([[1],[2],[3],[4]])` |
| Matrix | `A = [1,2;3,4]` | `np.array([[1,2],[3,4]])` |
| Index element | `A(2,3)` — 1-based | `A[1,2]` — 0-based |
| Monsoon slice | `v(6:9)` | `v[5:9]` |
| Last element | `v(end)` | `v[-1]` |
| Row sums | `sum(A,2)` | `A.sum(axis=1)` |
| Column means | `mean(A,1)` | `A.mean(axis=0)` |
| Size | `size(A,1)`, `size(A,2)` | `A.shape[0]`, `A.shape[1]` |
| For loop | `for i = 1:n ... end` | `for i in range(n):` |

Using the pipe network matrix from Code Block 4, extend the script to also compute:

````python
# Copy and run this in MATLAB Online
print("""
% Day38_Assignment.m
clc; clear;
pipes = [200,300,0.013,0.001; 150,250,0.013,0.002; 180,200,0.015,0.003;
         120,300,0.013,0.001;  90,200,0.015,0.002; 100,250,0.013,0.0015];

fprintf('%4s %6s %8s %10s %8s\n','Pipe','D(mm)','V(m/s)','Q(L/s)','Status')
for i = 1:size(pipes,1)
    D_mm=pipes(i,2); n=pipes(i,3); S=pipes(i,4);
    D=D_mm/1000; R=D/4; A=pi*(D/2)^2;
    V=(1/n)*R^(2/3)*S^0.5; Q=V*A;
    if V>=0.6 && V<=3.0, st='OK'; else, st='REVIEW'; end
    fprintf('%4d %6d %8.3f %10.2f %8s\n',i,D_mm,V,Q*1000,st)
end
""")
````
