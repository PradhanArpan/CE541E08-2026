## Code Block 1 — Froude Number and Hydraulic Jump

### What this code does

We compute the Froude number for a channel flow, classify it as subcritical or supercritical using if-else, and compute the sequent depth for a hydraulic jump if supercritical.

### Why each step is taken

**MATLAB if-else syntax:**
```matlab
if condition
    % true block
else
    % false block
end
```
No colon after `if`. The block ends with `end` (not indentation). Indentation is optional in MATLAB but good practice.

**`&&` for logical AND:**
MATLAB uses `&&` for short-circuit logical AND between scalars (same as Python's `and`). `&` is element-wise AND for arrays (same as Python/NumPy's `&`).

**Reynolds number check:**
`Re > 4000` produces a `logical` value (0 or 1). `%d` in fprintf prints it as an integer — 1 for turbulent, 0 for laminar.

### Algorithm

```
1. V=2.8, y=0.6, g=9.81
   Fr = V / sqrt(g*y)

2. if Fr < 1: print SUBCRITICAL
   else:
     y2 = (y/2)*(sqrt(1+8*Fr^2)-1)
     print SUPERCRITICAL and y2

3. Re = V*D/nu
   print Re and turbulence check
```

### Expected output

```
Froude number Fr = 1.153
Flow: SUPERCRITICAL - sequent depth = 1.375 m
Reynolds Re = 834329 - Turbulent: 1
```

````python
# Copy and run this in MATLAB Online
print("""
% CE541E08 - Day 39: Mathematical Expressions
V=2.8; y=0.6; g=9.81;
Fr = V / sqrt(g*y);
fprintf('Froude number Fr = %.3f\n', Fr)

if Fr < 1
    fprintf('Flow: SUBCRITICAL\n')
else
    y2 = (y/2)*(sqrt(1+8*Fr^2)-1);
    fprintf('Flow: SUPERCRITICAL - sequent depth = %.3f m\n', y2)
end

nu=1.007e-6; D=0.3;
Re = V*D/nu;
fprintf('Reynolds Re = %.0f - Turbulent: %d\n', Re, Re>4000)
""")
````

## Code Block 2 — Darcy-Weisbach for a Range of Velocities

### What this code does

We compute the D-W head loss and discharge for a range of velocities from 0.5 to 3.5 m/s using a for loop over a vector — the MATLAB equivalent of NumPy vectorised operations.

### Why each step is taken

**`V_array = 0.5:0.25:3.5`:**
Creates a row vector: [0.5, 0.75, 1.0, ..., 3.5]. The for loop then steps through each value.

**`V_array(i)` inside the loop:**
Access the i-th element of the vector. MATLAB for loops can also iterate directly: `for V = V_array` — then V takes each value in sequence without needing index i.

**`Q_array = V_array*pi*(D/2)^2`:**
Note: this uses scalar `pi*(D/2)^2` multiplied by the vector `V_array`. In MATLAB, multiplying a scalar by a vector applies to all elements. The result is a vector of Q values. No `.` is needed for scalar-vector multiplication.

### Algorithm

```
1. D=0.3, f=0.018, L=250, g=9.81
2. V_array = 0.5:0.25:3.5  (13 values)
3. h_f = f*(L/D)*(V^2/(2*g)) for each V
4. Q   = V*pi*(D/2)^2 for each V
5. Print 3-column table: V, Q(L/s), hf(m)
```

### Expected output

```
  V(m/s)     Q(L/s)      hf(m)
    0.50       3.53       0.191
    0.75       5.30       0.430
    1.00       7.07       0.765
    ...
    3.50      24.74       9.362
```

````python
# Copy and run this in MATLAB Online
print("""
% D-W head loss for a range of velocities
D=0.3; f=0.018; L=250; g=9.81;
V_array = 0.5:0.25:3.5;
h_f_array = f*(L/D)*(V_array.^2/(2*g));
Q_array   = V_array*pi*(D/2)^2;

fprintf('%8s %10s %12s\n','V(m/s)','Q(L/s)','hf(m)')
for i = 1:length(V_array)
    fprintf('%8.2f %10.2f %12.3f\n', V_array(i), Q_array(i)*1000, h_f_array(i))
end
""")
````

## Code Block 3 — SCS-CN for a Storm Series

### What this code does

We compute SCS-CN runoff for a range of storm depths from 10 to 150 mm using a for loop, and print a table showing runoff depth and runoff ratio for each storm.

### Why each step is taken

**`P_range = 10:10:150`:**
Storm depths from 10 to 150 mm in steps of 10 mm — 15 values. This is a typical sensitivity table in stormwater design.

**`Q_range = zeros(size(P_range))`:**
Initialises Q_range as a vector of zeros with the same size as P_range. Pre-allocating the output array before the loop is good MATLAB practice — it is faster than appending to an array in each iteration.

**`if P > Ia` inside the loop:**
Runoff only occurs when storm depth exceeds the initial abstraction. For storms smaller than Ia, Q remains 0.

### Algorithm

```
1. CN=75, S=25400/75-254=84.7, Ia=0.2*84.7=16.9 mm

2. P_range = 10:10:150 (15 storms)
   Q_range = zeros(1,15)   (pre-allocate)

3. For each P:
   if P > Ia:
     Q = (P-Ia)^2/(P-Ia+S)
   else: Q = 0

4. Print: P, Q, Q/P*100 (%)
```

### Expected output

```
CN=75, S=84.7 mm, Ia=16.9 mm
   P(mm)   Q(mm)   Q/P(%)
      10    0.00      0.0
      20    0.11      0.6
      30    1.47      4.9
      40    4.94     12.4
      ...
     150   82.20     54.8
```

````python
# Copy and run this in MATLAB Online
print("""
% SCS-CN runoff computation
CN=75; S=25400/CN-254; Ia=0.2*S;
P_range = 10:10:150;
Q_range = zeros(size(P_range));   % pre-allocate output

for i = 1:length(P_range)
    P = P_range(i);
    if P > Ia
        Q_range(i) = (P-Ia)^2/(P-Ia+S);
    end
end

fprintf('CN=%d, S=%.1f mm, Ia=%.1f mm\n', CN, S, Ia)
fprintf('%8s %8s %8s\n','P(mm)','Q(mm)','Q/P(%)')
for i = 1:length(P_range)
    pct = 0;
    if P_range(i) > 0, pct = Q_range(i)/P_range(i)*100; end
    fprintf('%8d %8.2f %8.1f\n', P_range(i), Q_range(i), pct)
end
""")
````

## Code Block 4 — Manning's Pipe Sizing Table (Nested Output)

### What this code does

We produce a full pipe sizing table: 7 diameters × 4 slopes. The outer loop iterates over diameters, the inner loop iterates over slopes — computing and printing Q (L/s) for each combination.

### Why each step is taken

**`for D_mm = diameters_mm`:**
MATLAB for loops can iterate directly over a vector. `D_mm` takes the value 150, then 200, then 250, ..., 600. No index needed.

**`for S = slopes`:**
Same pattern — S takes 0.001, then 0.002, then 0.003, then 0.005.

**`fprintf(' %9.2f', Q*1000)` with no `\n`:**
Printing without a newline keeps values on the same row. `fprintf('\n')` at the end of the outer loop moves to the next row.

### Algorithm

```
1. diameters_mm = [150,200,...,600]  (7 values)
   slopes = [0.001,0.002,0.003,0.005] (4 values)

2. Print header row with slope labels

3. For each D in diameters_mm:
   Print D at start of row
   For each S in slopes:
     Compute V and Q
     Print Q on same line (no newline)
   Print newline at end of row
```

### Expected output

```
D(mm) S=1:1000  S=1:500   S=1:333   S=1:200
  150     30.62     43.30     53.02     68.51
  200     61.17     86.49    105.88    136.82
  ...
  600    954.24   1349.43   1652.50   2134.93
```

````python
# Copy and run this in MATLAB Online
print("""
% Manning's pipe sizing table (7 diameters x 4 slopes)
diameters_mm = [150,200,250,300,375,450,600];
slopes = [0.001,0.002,0.003,0.005];
n = 0.013;

fprintf('%6s','D(mm)');
for S = slopes
    fprintf(' S=1:%-5d', round(1/S));
end
fprintf('\n');

for D_mm = diameters_mm
    D=D_mm/1000; R=D/4; A=pi*(D/2)^2;
    fprintf('%6d', D_mm);
    for S = slopes
        V=(1/n)*R^(2/3)*S^0.5; Q=V*A;
        fprintf(' %9.2f', Q*1000);
    end
    fprintf('\n')
end
""")
````

## Session Summary — Mathematical Expressions

| Concept | MATLAB | Python |
|---|---|---|
| If-else | `if cond ... else ... end` | `if cond: ... else: ...` |
| Logical AND (scalar) | `&&` | `and` |
| Logical OR (scalar) | `\|\|` | `or` |
| Cumulative sum | `cumsum(v)` | `np.cumsum(v)` |
| Pre-allocate | `zeros(size(v))` | `np.zeros(len(v))` |
| For over vector | `for x = vec_name` | `for x in vec_name:` |
| No newline | `fprintf('...')` without `\n` | `print(..., end='')` |
| Discard output | `[~, idx] = min(...)` | `_, idx = ...` |

## Day 39 Assignment

Write a MATLAB script that:
1. Uses a for loop to compute Manning's V and SCS-CN Q for the 6 Cauvery sub-catchments
2. For each catchment, prints: name, CN, S, Ia, Q(mm), runoff volume (Mm³)
3. Prints the basin total volume

````python
# Copy and run this in MATLAB Online
print("""
% Day39_Assignment.m
clc; clear;
names    = {'Hemavathi','Harangi','Kabini','Suvarnavathi','Shimsha','Arkavathi'};
area_km2 = [2950, 1930, 7040, 1575, 4762, 3101];
CN       = [72, 68, 75, 70, 65, 73];
P        = 85;   % storm depth, mm

fprintf('%-15s %4s %7s %7s %8s %10s\n','Catchment','CN','S(mm)','Ia(mm)','Q(mm)','Vol(Mm3)')
total_vol = 0;
for i = 1:length(CN)
    S  = 25400/CN(i) - 254;
    Ia = 0.2*S;
    Q  = 0;
    if P > Ia, Q = (P-Ia)^2/(P-Ia+S); end
    Vol = Q/1000*area_km2(i);
    total_vol = total_vol + Vol;
    fprintf('%-15s %4d %7.1f %7.1f %8.2f %10.3f\n', names{i},CN(i),S,Ia,Q,Vol)
end
fprintf('%-15s %4s %7s %7s %8s %10.3f\n','TOTAL','','','','',total_vol)
""")
````
