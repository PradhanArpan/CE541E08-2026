### ▶ Phase I Task 3 — Exploratory Data Analysis (EDA)

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt

AMS_data = {'Year':list(range(1994,2024)),
    'Peak_m3s':[4567,8934,3456,12340,6789,9012,5678,15670,8901,4567,
                11234,7890,5678,9012,13456,6789,4567,10234,8901,5678,
                14567,7890,6789,11234,9012,5678,8901,12345,7890,6789]}
df = pd.DataFrame(AMS_data)

# Time series plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].bar(df['Year'], df['Peak_m3s'], color='steelblue', alpha=0.8)
axes[0].axhline(df['Peak_m3s'].mean(), color='red', linestyle='--',
                label=f"Mean: {df['Peak_m3s'].mean():.0f} m3/s")
axes[0].set_title('Annual Maximum Flood Series')
axes[0].set_xlabel('Year'); axes[0].set_ylabel('Peak Q (m3/s)')
axes[0].legend()

axes[1].hist(df['Peak_m3s'], bins=8, color='steelblue', edgecolor='white')
axes[1].axvline(df['Peak_m3s'].mean(), color='red', linestyle='--', label='Mean')
axes[1].set_title('Frequency Distribution of Peak Flows')
axes[1].set_xlabel('Peak Q (m3/s)'); axes[1].set_ylabel('Count')
axes[1].legend()

plt.tight_layout(); plt.show()

print("EDA Complete. Key observations:")
print(f"  CV (variability)  : {df['Peak_m3s'].std()/df['Peak_m3s'].mean():.3f}")
print(f"  Skewness          : {df['Peak_m3s'].skew():.3f}")
print(f"  Decade 1994-2003  : {df[df['Year']<=2003]['Peak_m3s'].mean():.0f} m3/s mean")
print(f"  Decade 2014-2023  : {df[df['Year']>=2014]['Peak_m3s'].mean():.0f} m3/s mean")
```
