## Code Block 1 — Comparison Operators

### What this code does

We evaluate six comparison operators on engineering values and combine them into compound design checks.

### Why each step is taken

**`velocity > 3.0`:** Returns True or False. The result is a boolean value.

**`diameter == 300`:** `==` is equality comparison. Do not confuse with `=` (assignment).

**`Q_actual > Q_design`:** Comparing two float variables — returns True if actual exceeds design capacity.

**`vel_ok = velocity >= 0.6 and velocity <= 3.0`:** Compound condition: velocity must satisfy both bounds. Python also allows `0.6 <= velocity <= 3.0` as a chained comparison.

### Algorithm

```
1. velocity=2.4, diameter=300, Q_actual=0.045, Q_design=0.040
2. Six individual comparisons printed
3. Compound condition: vel_ok = (0.6 <= V <= 3.0)
4. cap_ok = Q_actual >= Q_design
```

### Expected output

```
velocity > 3.0      : False
velocity >= 0.6     : True
diameter == 300     : True
Q_actual > Q_design : True

vel_ok (0.6-3.0)    : True
cap_ok (Q>=Qd)      : True
both OK             : True
```

````python
velocity = 2.4; diameter = 300; Q_actual = 0.045; Q_design = 0.040

print("Condition evaluations:")
print(f"  velocity > 3.0      : {velocity > 3.0}")
print(f"  velocity >= 0.6     : {velocity >= 0.6}")
print(f"  diameter == 300     : {diameter == 300}")
print(f"  Q_actual > Q_design : {Q_actual > Q_design}")

vel_ok  = velocity >= 0.6 and velocity <= 3.0
cap_ok  = Q_actual >= Q_design
print()
print(f"  vel_ok (0.6-3.0)    : {vel_ok}")
print(f"  cap_ok (Q>=Qd)      : {cap_ok}")
print(f"  both OK             : {vel_ok and cap_ok}")
````

## Code Block 2 — Simple if Block

### What this code does

We check whether a proposed pipe diameter meets the minimum code requirement, printing a margin if it passes.

### Why each step is taken

**`if proposed_mm >= min_mm:`:** Only the indented block runs when True. If False, the block is skipped entirely.

**`proposed_mm - min_mm`:** The margin above the minimum. Useful to print alongside the check — tells the engineer how much safety margin exists.

### Algorithm

```
1. proposed_mm=150, min_mm=100
2. if 150 >= 100:       → True
     print pass + margin
3. "Check complete." always runs
```

### Expected output

```
Diameter 150 mm meets minimum requirement.
Margin: 50 mm above minimum
Check complete.
```

````python
proposed_mm = 150   # try: 80, 100, 150, 300
min_mm      = 100

if proposed_mm >= min_mm:
    print(f"Diameter {proposed_mm} mm meets minimum requirement.")
    print(f"Margin: {proposed_mm - min_mm} mm above minimum")

print("Check complete.")
````

## Code Block 3 — if-else: Froude Number Classification

### What this code does

We compute the Froude number for a channel flow and classify it as subcritical or supercritical using if-else.

### Why each step is taken

**`if Fr < 1.0:` / `else:`:** The else block runs when the if condition is False. Exactly one of the two blocks executes — never both.

**Physical meaning of Fr:**
- Fr < 1: subcritical (tranquil flow) — disturbances can propagate upstream
- Fr > 1: supercritical (shooting flow) — disturbances cannot propagate upstream

### Algorithm

```
1. V=3.5, y=0.4, g=9.81
2. Fr = 3.5 / sqrt(9.81 × 0.4) = 1.767
3. if 1.767 < 1.0:  → False → else block runs
4. Print SUPERCRITICAL
```

### Expected output

```
V=3.5 m/s, y=0.4 m, Fr=1.767
SUPERCRITICAL (Fr > 1) — shooting flow
  Disturbances cannot propagate upstream.
```

````python
import math
V=3.5; y=0.4; g=9.81
Fr = V / math.sqrt(g * y)

print(f"V={V} m/s, y={y} m, Fr={Fr:.3f}")

if Fr < 1.0:
    print("SUBCRITICAL (Fr < 1) — tranquil flow")
    print("  Disturbances propagate upstream.")
else:
    print("SUPERCRITICAL (Fr > 1) — shooting flow")
    print("  Disturbances cannot propagate upstream.")
````

## Code Block 4 — if-else: IMD 2-Category Rainfall Classification

### What this code does

We classify daily rainfall as Heavy-or-above vs Light/Moderate using the IMD threshold of 64.5 mm, then compute runoff volume.

### Why each step is taken

**`if rainfall_mm >= 64.5:`:** The official IMD threshold for Heavy Rain category. Any day at or above this triggers Orange/Red alert protocols.

**`volume_Mm3 = rainfall_mm/1000 * catchment_km2*1e6 / 1e6`:** Converting rainfall depth (mm) to volume (Mm³):
- `/1000` converts mm to m
- `*catchment_km2*1e6` converts km² to m², giving m³
- `/1e6` converts m³ to Mm³ (million cubic metres)

### Algorithm

```
1. rainfall_mm=87.3, catchment_km2=1250.0
2. 87.3 >= 64.5 → True → HEAVY block
3. volume = 87.3/1000 × 1250 × 1e6 / 1e6 = 109.125 Mm³
```

### Expected output

```
Daily rainfall: 87.3 mm
Classification: HEAVY or above (>= 64.5 mm)
  Check upstream gauges — IMD Orange/Red Alert protocol.

  Potential runoff on 1250.0 km²: 109.1 Mm3
```

````python
rainfall_mm   = 87.3   # try: 10, 45, 65, 120, 210
catchment_km2 = 1250.0

print(f"Daily rainfall: {rainfall_mm} mm")

if rainfall_mm >= 64.5:
    print("Classification: HEAVY or above (≥ 64.5 mm)")
    print("  Check upstream gauges — IMD Orange/Red Alert protocol.")
else:
    print("Classification: LIGHT or MODERATE (< 64.5 mm)")
    print("  Routine monitoring continues.")

volume_Mm3 = rainfall_mm/1000 * catchment_km2*1e6 / 1e6
print()
print(f"  Potential runoff on {catchment_km2} km²: {volume_Mm3:.1f} Mm3")
````

## Session Summary — Comparison Operators and if-else

| Concept | Syntax | Notes |
|---|---|---|
| Equality | `a == b` | Not `=` (which is assignment) |
| Not equal | `a != b` | |
| Compound | `a >= 0.6 and a <= 3.0` | or `0.6 <= a <= 3.0` |
| Simple if | `if cond: ...` | Block skipped if False |
| if-else | `if cond: ... else: ...` | One branch always runs |
| Froude | `Fr = V/sqrt(g*y)` | `<1` subcritical, `>1` supercritical |
| IMD heavy | `rain >= 64.5` | Threshold for Heavy Rain alert |
