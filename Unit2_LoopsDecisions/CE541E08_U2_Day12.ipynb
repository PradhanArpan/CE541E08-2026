## Section 1 — elif: the multi-way decision

When you have more than two outcomes, use `elif` (short for "else if"):

```python
if condition_1:
    block_1()
elif condition_2:
    block_2()
elif condition_3:
    block_3()
else:
    block_default()
```

Python checks each condition **in order, top to bottom**.
The **first True condition** runs its block — all others are skipped.
`else` catches everything that did not match above.

### ▶ Cell 1 — Full IMD 6-category rainfall classification

````python
# ---------------------------------------------------------
# WHAT THIS CELL DOES:
# Implements the complete IMD daily rainfall classification
# using if-elif-else with all 6 official categories.
# This is used by hydrologists across India every monsoon day.
# ---------------------------------------------------------

rainfall_mm = 156.2   # ← try: 0, 5, 30, 70, 140, 210

# IMD official thresholds (mm/day)
if rainfall_mm == 0:
    category = "No rain"
    alert    = "—"
    action   = "No action required."

elif rainfall_mm <= 15.5:
    category = "Light rain"
    alert    = "Green"
    action   = "Routine monitoring."

elif rainfall_mm <= 64.4:
    category = "Moderate rain"
    alert    = "Yellow"
    action   = "Check reservoir inflows."

elif rainfall_mm <= 115.5:
    category = "Heavy rain"
    alert    = "Orange"
    action   = "Alert dam operators. Monitor upstream."

elif rainfall_mm <= 204.4:
    category = "Very Heavy rain"
    alert    = "Red"
    action   = "Open spillway gates. Issue flood watch."

else:
    category = "Extremely Heavy rain"
    alert    = "Red (Extreme)"
    action   = "FLOOD WARNING. Evacuate low-lying areas."

print(f"Rainfall          : {rainfall_mm} mm/day")
print(f"IMD Category      : {category}")
print(f"Alert colour      : {alert}")
print(f"Recommended action: {action}")

# --- INSTRUCTOR NOTE ---------------------------------------
# CLASSROOM EXERCISE: Run for each of these values:
# 0 → No rain, 5 → Light, 30 → Moderate,
# 70 → Heavy, 140 → Very Heavy, 210 → Extremely Heavy
# Ask: "Which threshold do dam safety authorities watch most?"
# Answer: 115.5 mm — Very Heavy triggers spillway decisions.
# -----------------------------------------------------------
````

## Section 2 — Nested if-else: two-variable classification

Sometimes the decision depends on **two variables at once** — 
a nested if inside an outer if.

### ▶ Cell 2 — Soil erosion risk: rainfall × slope class

````python
# ---------------------------------------------------------
# WHAT THIS CELL DOES:
# Classifies soil erosion risk based on BOTH rainfall
# intensity AND land slope — a real watershed management
# assessment used in India's National Watershed Programme.
# ---------------------------------------------------------

rainfall_mm  = 87.0   # mm/day  ← try combinations
slope_pct    = 18.0   # %       ← try: 3, 12, 18, 30

# Classify rainfall (simplified: low / high)
if rainfall_mm >= 64.5:
    rain_class = "high"
else:
    rain_class = "low"

# Classify slope (gentle <10%, moderate 10-25%, steep >25%)
if slope_pct < 10:
    slope_class = "gentle"
elif slope_pct <= 25:
    slope_class = "moderate"
else:
    slope_class = "steep"

# Nested decision: erosion risk from both factors
if rain_class == "high":
    if slope_class == "steep":
        risk = "VERY HIGH"
        measure = "Immediate terracing + vegetative cover required."
    elif slope_class == "moderate":
        risk = "HIGH"
        measure = "Contour bunding + grass waterways recommended."
    else:
        risk = "MODERATE"
        measure = "Check dams and mulching advised."
else:
    if slope_class == "steep":
        risk = "MODERATE"
        measure = "Preventive terracing recommended."
    else:
        risk = "LOW"
        measure = "Standard agricultural practices sufficient."

print(f"Rainfall       : {rainfall_mm} mm/day ({rain_class})")
print(f"Slope          : {slope_pct}% ({slope_class})")
print(f"Erosion risk   : {risk}")
print(f"Recommendation : {measure}")

# --- INSTRUCTOR NOTE ---------------------------------------
# This is a real decision matrix from NBSS&LUP guidelines.
# Try: rainfall=87, slope=30 → VERY HIGH (worst case)
# Try: rainfall=20, slope=5  → LOW (best case)
# Ask students to complete the full 3×2 matrix by trying
# all combinations. Write the matrix on the board.
# -----------------------------------------------------------
````

### ▶ Cell 3 — Manning's channel design: three-level velocity check

````python
import math

# ---------------------------------------------------------
# WHAT THIS CELL DOES:
# Gives a three-level design verdict for channel velocity:
# too slow (silting) / acceptable / too fast (erosion)
# ---------------------------------------------------------

# Channel parameters
B=3.0; y=1.0; n=0.025; S=0.0005
A = B*y; P = B+2*y; R = A/P
V = (1/n)*R**(2/3)*S**0.5
Q = V*A

print(f"Channel: B={B}m, y={y}m, n={n}, S={S}")
print(f"Velocity V = {V:.3f} m/s")
print()

if V < 0.5:
    verdict = "TOO SLOW — silting risk"
    advice  = f"Increase slope or reduce depth. Need +{0.5-V:.3f} m/s."
elif V <= 2.5:
    verdict = "ACCEPTABLE"
    advice  = f"Velocity within 0.5–2.5 m/s design range."
    margin  = ((2.5-V)/(2.5-0.5))*100
    advice += f" {margin:.0f}% below maximum."
else:
    verdict = "TOO FAST — erosion risk"
    advice  = f"Reduce slope or line channel with hard surface. Excess: {V-2.5:.3f} m/s."

print(f"Design verdict : {verdict}")
print(f"Recommendation : {advice}")
print(f"Discharge Q    : {Q:.4f} m³/s")

# --- INSTRUCTOR NOTE ---------------------------------------
# Three-level verdict: too slow / OK / too fast
# This is more useful than a binary pass/fail.
# Ask: "What S would give exactly 0.5 m/s for this channel?"
# S = (V*n / R^(2/3))^2 = (0.5*0.025 / 0.6^(2/3))^2 ≈ 0.000139
# -----------------------------------------------------------
````

### ▶ Cell 4 — Water quality index: multi-parameter classification

````python
# ---------------------------------------------------------
# WHAT THIS CELL DOES:
# Classifies drinking water quality based on pH and turbidity
# using nested if-elif-else — a simplified WQI check.
# ---------------------------------------------------------

pH        = 7.2    # ← try: 5.5, 6.2, 7.2, 8.8, 9.5
turbidity = 3.5    # NTU  ← try: 1, 4, 12, 25

# BIS IS:10500 drinking water standards
if pH < 6.5 or pH > 8.5:
    ph_status = "OUT OF RANGE"
    ph_ok = False
else:
    ph_status = "acceptable"
    ph_ok = True

if turbidity <= 1.0:
    turb_status = "excellent"
    turb_ok = True
elif turbidity <= 5.0:
    turb_status = "acceptable"
    turb_ok = True
elif turbidity <= 10.0:
    turb_status = "requires treatment"
    turb_ok = False
else:
    turb_status = "UNSAFE — high turbidity"
    turb_ok = False

# Overall verdict
if ph_ok and turb_ok:
    overall = "FIT FOR DRINKING ✓"
elif ph_ok or turb_ok:
    overall = "REQUIRES TREATMENT before drinking"
else:
    overall = "UNFIT — multiple parameters out of range"

print(f"pH          : {pH} ({ph_status})")
print(f"Turbidity   : {turbidity} NTU ({turb_status})")
print(f"Overall     : {overall}")

# --- INSTRUCTOR NOTE ---------------------------------------
# BIS IS:10500 standard for drinking water.
# pH 6.5-8.5, turbidity ≤ 1 NTU desirable, ≤ 5 permissible.
# The 'or' in the overall verdict is important —
# ask students: "When is water 'requires treatment'?"
# Answer: when exactly one parameter fails (not both).
# -----------------------------------------------------------
````
