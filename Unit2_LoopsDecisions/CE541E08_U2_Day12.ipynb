## Section 1 — Multi-Way Decisions with if-elif-else

When a value falls into more than two categories, use `elif` (else-if). Python checks conditions top to bottom and stops at the first True. Only one branch executes.

```python
if condition_1:
    # first choice
elif condition_2:
    # second choice
elif condition_3:
    # third choice
else:
    # default (none of the above)
```

The `else` is optional but recommended — it handles unexpected values.

## Code Block 1 — IMD 6-Category Rainfall Classification

### What this code does

We implement the complete Indian Meteorological Department (IMD) rainfall classification system with six official categories, matching exactly the thresholds used in daily weather bulletins across India.

### Why each step is taken

**Why conditions are ordered from lowest to highest:**
Python checks conditions in order and stops at the first True. If we checked `rainfall_mm < 204.4` before `rainfall_mm < 115.5`, every Heavy Rain reading would incorrectly be classified as Very Heavy. The specific category must be checked before the more general one.

**Why each elif is needed:**
Each `elif` only runs if all previous conditions were False. So when we reach `elif rainfall_mm <= 115.5`, we already know `rainfall_mm > 64.4`. Together they define the Heavy Rain interval.

**`alert` and `action` variables:**
Storing the alert level and recommended action in variables before printing allows easy reuse — for example, sending the same values to a database or SMS alert system.

### Algorithm

```
1. rainfall_mm = 156.2

2. if == 0:           → No rain
   elif <= 15.5:      → Light
   elif <= 64.4:      → Moderate
   elif <= 115.5:     → Heavy (64.5 to 115.5)
   elif <= 204.4:     → Very Heavy (115.6 to 204.4)
   else:              → Extremely Heavy (> 204.4)

3. Print category, alert, action, runoff volume
```

### Expected output (rainfall=156.2mm)

```
Daily rainfall    : 156.2 mm
IMD category      : Very Heavy Rain
Alert level       : Red Alert
Recommended action: Issue dam safety advisory. Activate downstream warnings.
Runoff potential  : 156.2 Mm3 on 1000 km²
```

````python
rainfall_mm = 156.2   # try: 0, 5, 30, 70, 140, 210

# IMD official thresholds (mm/day)
if rainfall_mm == 0:
    category = "No rain";           alert = "—";           action = "No action"
elif rainfall_mm <= 15.5:
    category = "Light Rain";        alert = "Green";       action = "Routine monitoring"
elif rainfall_mm <= 64.4:
    category = "Moderate Rain";     alert = "Yellow";      action = "Monitor river levels"
elif rainfall_mm <= 115.5:
    category = "Heavy Rain";        alert = "Orange Alert"; action = "Alert reservoir operators"
elif rainfall_mm <= 204.4:
    category = "Very Heavy Rain";   alert = "Red Alert";   action = "Issue dam safety advisory. Activate downstream warnings."
else:
    category = "Extremely Heavy";   alert = "Red Alert+++"; action = "Emergency evacuation protocol"

print(f"Daily rainfall    : {rainfall_mm} mm")
print(f"IMD category      : {category}")
print(f"Alert level       : {alert}")
print(f"Recommended action: {action}")
runoff_Mm3 = rainfall_mm / 1000 * 1000 * 1e6 / 1e6   # 1000 km² catchment
print(f"Runoff potential  : {runoff_Mm3:.1f} Mm3 on 1000 km²")
````

## Code Block 2 — Nested if: Soil Erosion Risk

### What this code does

We assess soil erosion risk based on both rainfall intensity and land slope using a nested if-elif structure — the same logic used in India's National Watershed Programme.

### Why each step is taken

**Nested conditions:**
The outer if-elif classifies rainfall into three levels. Inside each branch, an inner if-elif further classifies slope. Together they produce a 3×3 risk matrix: 9 combinations give 4 distinct risk levels.

**Why nesting instead of `and`:**
For complex multi-variable decisions, nested ifs can be clearer than long `and` chains. The structure mirrors a decision tree that a field engineer would follow.

### Algorithm

```
1. rainfall=87.0, slope=18.0

2. Outer: classify rainfall
   if >= 64.5:   rain_class = "High"
   elif >= 15.5: rain_class = "Moderate"
   else:         rain_class = "Low"

3. Inner (within High rain branch):
   if slope >= 20: risk = "VERY HIGH"
   elif slope >= 10: risk = "HIGH"
   else:             risk = "MODERATE"
```

### Expected output (rainfall=87.0, slope=18.0)

```
Rainfall  : 87.0 mm (High)
Slope     : 18.0%
EROSION RISK: HIGH
  Action: Contour bunding required. Vegetative barriers recommended.
```

````python
rainfall_mm  = 87.0   # try combinations
slope_pct    = 18.0   # %

if rainfall_mm >= 64.5:
    rain_class = "High"
    if slope_pct >= 20:
        risk = "VERY HIGH"; action = "Immediate engineering intervention required"
    elif slope_pct >= 10:
        risk = "HIGH";      action = "Contour bunding required. Vegetative barriers recommended."
    else:
        risk = "MODERATE";  action = "Periodic monitoring. Check drainage outlets."
elif rainfall_mm >= 15.5:
    rain_class = "Moderate"
    if slope_pct >= 20:
        risk = "HIGH";      action = "Slope stabilisation recommended"
    else:
        risk = "LOW";       action = "Standard agricultural practices"
else:
    rain_class = "Low"
    risk = "LOW";           action = "No special action needed"

print(f"Rainfall  : {rainfall_mm} mm ({rain_class})")
print(f"Slope     : {slope_pct}%")
print(f"EROSION RISK: {risk}")
print(f"  Action: {action}")
````

## Code Block 3 — Three-Level Velocity Verdict

### What this code does

We give a three-level design verdict for channel velocity: too slow (silting risk), acceptable, or too fast (erosion risk) — computed from actual Manning's formula.

### Why each step is taken

**Compute V from formula, then classify:**
This is more realistic than using a hard-coded V — the classification depends on a computed result, which is the typical workflow.

**Three conditions:**
- `V < 0.5`: below self-cleansing (silting risk)
- `0.5 <= V < 2.5`: acceptable range for earthen channels
- `V >= 2.5`: above erosion threshold

### Expected output

```
Channel: B=3.0m, y=1.0m, n=0.025, S=0.0005
Velocity V = 0.447 m/s

VERDICT: TOO SLOW (< 0.5 m/s)
  Risk: Sediment deposition — drainage channel will block.
  Fix: Increase slope or reduce cross-section.
```

````python
import math

# Channel parameters
B=3.0; y=1.0; n=0.025; S=0.0005
A = B*y; P = B+2*y; R = A/P
V = (1/n)*R**(2/3)*S**0.5
Q = V*A

print(f"Channel: B={B}m, y={y}m, n={n}, S={S}")
print(f"Velocity V = {V:.3f} m/s")
print()

if V < 0.5:
    verdict = "TOO SLOW (< 0.5 m/s)"
    detail  = "Risk: Sediment deposition — drainage channel will block.
  Fix: Increase slope or reduce cross-section."
elif V <= 2.5:
    verdict = "ACCEPTABLE (0.5 – 2.5 m/s)"
    detail  = f"Design OK. Q = {Q:.3f} m³/s ({Q*1000:.1f} L/s)"
else:
    verdict = "TOO FAST (> 2.5 m/s)"
    detail  = "Risk: Bed and bank erosion.
  Fix: Reduce slope or increase cross-section."

print(f"VERDICT: {verdict}")
print(f"  {detail}")
````

## Session Summary — if-elif-else

| Pattern | When to use | Key point |
|---|---|---|
| `if-else` | 2 outcomes | Exactly one branch runs |
| `if-elif-else` | 3+ outcomes | First True wins; else is default |
| Nested if | Multi-variable decisions | Outer condition sets context |
| Order matters | Specific before general | `elif <= 64.4` before `elif <= 115.5` |

## Day 12 Assignment

Three catchments — classify each for erosion risk and print a summary table.

### ▶ Assignment cell

````python
catchments = [
    ("Catchment A", 480,  3),
    ("Catchment B", 950,  12),
    ("Catchment C", 1450, 22),
]

for name, rainfall_mm_yr, slope_pct in catchments:
    # Classify rainfall (approximate annual → daily: divide by 365)
    daily_approx = rainfall_mm_yr / 365
    if daily_approx >= 3.0:
        rain_class = "High"
    elif daily_approx >= 1.0:
        rain_class = "Medium"
    else:
        rain_class = "Low"

    # Classify slope
    if slope_pct < 5:
        slope_class = "Flat"
    elif slope_pct < 15:
        slope_class = "Rolling"
    else:
        slope_class = "Hilly"

    # Erosion risk (combined)
    if rain_class == "High" and slope_class == "Hilly":
        risk = "VERY HIGH"
    elif rain_class in ("High","Medium") and slope_class in ("Rolling","Hilly"):
        risk = "HIGH"
    else:
        risk = "LOW"

    print(f"{name:<15} Rain={rain_class:<6} Slope={slope_class:<8} Risk={risk}")
````
