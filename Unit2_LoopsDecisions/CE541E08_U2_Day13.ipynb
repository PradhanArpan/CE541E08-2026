## Section 1 — The for loop

A `for` loop repeats a block of code **for each item in a sequence**:

```python
for item in sequence:
    # this block runs once for each item
    do_something(item)
```

**Without a loop:**
```python
print(rainfall[0])
print(rainfall[1])
print(rainfall[2])   # tedious for 30 days, impossible for 30 years
```

**With a loop:**
```python
for rain in rainfall:
    print(rain)      # works for 30 days or 30 years — same code
```

### ▶ Cell 1 — for loop over a daily rainfall list

````python
# ---------------------------------------------------------
# WHAT THIS CELL DOES:
# Iterates over daily rainfall data using a for loop.
# Computes cumulative total, identifies rainy days, finds peak.
# ---------------------------------------------------------

# June 2024 daily rainfall, Bengaluru (mm) — hypothetical
june_rainfall = [
    0.0, 0.0, 12.4, 45.6, 0.0, 8.2, 23.1,
    0.0, 0.0, 87.3, 34.5, 0.0, 56.2, 0.0,
    18.9, 0.0, 0.0, 134.5, 22.3, 45.6,
    0.0, 0.0, 67.8, 12.1, 0.0, 89.4, 33.2,
    0.0, 45.1, 28.7
]

total_mm    = 0.0
rainy_days  = 0
peak_rain   = 0.0
peak_day    = 0

for i, rain in enumerate(june_rainfall, start=1):
    total_mm   += rain
    if rain > 0:
        rainy_days += 1
    if rain > peak_rain:
        peak_rain = rain
        peak_day  = i

print(f"June 2024 Rainfall Analysis — Bengaluru")
print(f"{'-'*40}")
print(f"Total rainfall   : {total_mm:.1f} mm")
print(f"Monthly mean     : {total_mm/len(june_rainfall):.2f} mm/day")
print(f"Rainy days       : {rainy_days} of {len(june_rainfall)}")
print(f"Dry days         : {len(june_rainfall)-rainy_days}")
print(f"Peak rainfall    : {peak_rain} mm on Day {peak_day}")
print(f"Peak IMD class   : {'Very Heavy' if peak_rain>115.5 else 'Heavy' if peak_rain>64.5 else 'Moderate'}")

# --- INSTRUCTOR NOTE ---------------------------------------
# enumerate(list, start=1) gives (index, value) pairs.
# The loop does in one pass what Unit 4 (Pandas) will do
# with a single function call on 30 years of data.
# Ask: "What change gives the mean only for rainy days?"
# Answer: divide total_mm by rainy_days instead of len().
# -----------------------------------------------------------
````

### ▶ Cell 2 — range() for pipe section analysis

````python
# ---------------------------------------------------------
# WHAT THIS CELL DOES:
# Uses range() to iterate over a pipeline with multiple
# sections, computing cumulative head loss along the route.
# ---------------------------------------------------------

import math

# Pipeline data: 6 sections
diameters_mm = [300, 300, 250, 250, 200, 200]
lengths_m    = [150, 200, 180, 120, 90,  100]
velocity_ms  = [1.2, 1.2, 1.8, 1.8, 2.8, 2.8]
f_darcy      = 0.018
g            = 9.81

total_head_loss = 0.0
print(f"{'Sec':>4} {'D(mm)':>6} {'L(m)':>6} {'V(m/s)':>8} {'hf(m)':>8} {'Cumul(m)':>10}")
print("-" * 48)

for i in range(len(diameters_mm)):
    D   = diameters_mm[i] / 1000
    L   = lengths_m[i]
    V   = velocity_ms[i]
    h_f = f_darcy * (L/D) * (V**2 / (2*g))
    total_head_loss += h_f
    print(f"{i+1:>4} {diameters_mm[i]:>6} {L:>6} {V:>8.2f} {h_f:>8.3f} {total_head_loss:>10.3f}")

print("-" * 48)
print(f"Total head loss along pipeline: {total_head_loss:.3f} m")

# --- INSTRUCTOR NOTE ---------------------------------------
# range(len(list)) gives indices 0,1,2,...,n-1.
# This is the manual version of what zip() does — shown Day 14.
# Ask: "How do you find the section with the highest head loss?"
# Hint: track max h_f and its index — done in next session.
# -----------------------------------------------------------
````

### ▶ Cell 3 — Nested for loop: rainfall-runoff matrix

````python
# ---------------------------------------------------------
# WHAT THIS CELL DOES:
# Uses nested for loops to compute SCS-CN runoff for
# multiple combinations of rainfall and catchment type.
# ---------------------------------------------------------

rainfalls_mm = [20, 40, 60, 80, 100]     # design storms
cn_values    = [60, 70, 80, 90]          # curve numbers

print("SCS-CN Runoff Depth (mm)")
print(f"{'CN':>4}", end="")
for P in rainfalls_mm:
    print(f"  P={P:>3}mm", end="")
print()
print("-" * 52)

for CN in cn_values:
    S = 25400/CN - 254       # potential max retention, mm
    print(f"{CN:>4}", end="")
    for P in rainfalls_mm:
        Ia = 0.2 * S         # initial abstraction
        if P > Ia:
            Q = (P - Ia)**2 / (P - Ia + S)
        else:
            Q = 0.0
        print(f"  {Q:>7.2f}", end="")
    print()

# --- INSTRUCTOR NOTE ---------------------------------------
# SCS-CN method: Q = (P-Ia)²/(P-Ia+S), Ia=0.2S, S=25400/CN-254
# Nested loops: outer = CN (4 iterations), inner = P (5 iterations)
# Total = 20 computations in 6 lines of code.
# Ask: "How many hand calculations would this replace?"
# Answer: 20 — a full design storm matrix in seconds.
# -----------------------------------------------------------
````

### ▶ Cell 4 — for loop with if inside: classify all days

````python
# ---------------------------------------------------------
# WHAT THIS CELL DOES:
# Combines for loop + if-elif-else to classify every day
# in a month's rainfall record simultaneously.
# ---------------------------------------------------------

june_rainfall = [
    0.0, 0.0, 12.4, 45.6, 0.0, 8.2, 23.1,
    0.0, 0.0, 87.3, 34.5, 0.0, 56.2, 0.0,
    18.9, 0.0, 0.0, 134.5, 22.3, 45.6,
    0.0, 0.0, 67.8, 12.1, 0.0, 89.4, 33.2,
    0.0, 45.1, 28.7
]

counts = {"No rain":0,"Light":0,"Moderate":0,"Heavy":0,"Very Heavy":0,"Extremely Heavy":0}

for day_num, rain in enumerate(june_rainfall, start=1):
    if rain == 0:
        cat = "No rain"
    elif rain <= 15.5:
        cat = "Light"
    elif rain <= 64.4:
        cat = "Moderate"
    elif rain <= 115.5:
        cat = "Heavy"
    elif rain <= 204.4:
        cat = "Very Heavy"
    else:
        cat = "Extremely Heavy"
    counts[cat] += 1

print("June 2024 — IMD Rainfall Category Summary")
print(f"{'Category':<20} {'Days':>6} {'Bar'}")
print("-"*45)
for cat, count in counts.items():
    bar = "█" * count
    print(f"{cat:<20} {count:>6}  {bar}")

# --- INSTRUCTOR NOTE ---------------------------------------
# This is the exact pattern Unit 4 (Pandas) will do with
# pd.cut() on 30 years of data in one line.
# The manual version builds intuition for what Pandas does.
# -----------------------------------------------------------
````
