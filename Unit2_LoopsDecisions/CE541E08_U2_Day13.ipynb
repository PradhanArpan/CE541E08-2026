## Section 1 — The for Loop

A for loop repeats a block of code for each item in a sequence. In Python:

```python
for variable in sequence:
    # body — indented block
```

The sequence can be a list, a `range()`, or any iterable. Two key patterns for engineering data:

- **`enumerate(list, 1)`** — gives both value AND index (starting from 1), useful for day-numbered reports
- **`range(n)`** — generates integers 0 to n-1; `range(1,n+1)` generates 1 to n

## Code Block 1 — for Loop with enumerate: June Rainfall

### What this code does

We iterate over June's 30-day rainfall record, compute the running cumulative total, identify Heavy Rain days, and find the peak.

### Why each step is taken

**`for i, rain in enumerate(june_rainfall, 1)`:**
`enumerate` produces pairs of (index, value). The second argument `1` sets the starting index — so `i` goes 1, 2, 3, ... (day numbers, not 0-based).

**`cumulative += rain`:**
The `+=` operator adds `rain` to the running total after each day. Without this, we would have to recompute the sum from scratch each day.

**`if rain >= 64.5`:**
The IMD Heavy Rain threshold. Each day is classified inside the loop — this is for+if combined.

**`if rain > peak`:**
Updates the peak whenever a higher value is found. `peak_day` stores the corresponding day number.

### Algorithm

```
1. june_rainfall = [30 values]
   cumulative=0, heavy=0, peak=0, peak_day=0

2. for i, rain in enumerate(june_rainfall, 1):
     cumulative += rain
     if rain >= 64.5: heavy += 1
     if rain > peak: peak=rain; peak_day=i
     if i <= 8: print daily report row

3. After loop: print totals
```

### Expected output

```
Day   Rainfall(mm)  Cumulative  Class
──────────────────────────────────────
  1       0.0           0.0    No rain
  2       0.0           0.0    No rain
  3      12.4          12.4    Light
  ...
  8       0.0          89.3    No rain

Monthly: total=865.5mm, heavy=6 days, peak=134.5mm Day18
```

````python
june_rainfall = [
    0.0, 0.0, 12.4, 45.6, 0.0, 8.2, 23.1,
    0.0, 0.0, 87.3, 34.5, 0.0, 56.2, 0.0,
    18.9, 0.0, 0.0, 134.5, 22.3, 45.6,
    0.0, 0.0, 67.8, 12.1, 0.0, 89.4, 33.2,
    0.0, 45.1, 28.7
]

cumulative = 0.0; heavy = 0; peak = 0.0; peak_day = 0

print(f"{'Day':<5} {'Rainfall(mm)':<14} {'Cumulative':<12} {'Class'}")
print("─"*46)

for i, rain in enumerate(june_rainfall, 1):
    cumulative += rain
    if rain == 0:
        cls = "No rain"
    elif rain <= 15.5:
        cls = "Light"
    elif rain <= 64.4:
        cls = "Moderate"
    else:
        cls = "Heavy+"
        heavy += 1
    if rain > peak:
        peak = rain; peak_day = i
    if i <= 8:                     # print first 8 days
        print(f"{i:<5} {rain:<14.1f} {cumulative:<12.1f} {cls}")

print(f"
Monthly: total={cumulative:.1f}mm, heavy={heavy} days, peak={peak}mm Day{peak_day}")
````

## Code Block 2 — for with range: Pipeline Head Loss

### What this code does

We compute cumulative head loss along a 6-section pipeline using `range()` — accessing list elements by index and accumulating head loss from section to section.

### Why each step is taken

**`for i in range(6)`:**
`range(6)` generates 0, 1, 2, 3, 4, 5 — the six valid indices for the six-element lists. `i` is the index used to access `diameters_mm[i]`, `lengths_m[i]`, etc.

**Why `range()` instead of direct list iteration:**
When we need the index `i` to access multiple parallel lists simultaneously, `range()` is the clearest approach. If we only had one list, `for d in diameters_mm` would be cleaner.

**`total_head_loss += h_f`:**
Running accumulation — adds this section's head loss to the total. After 6 iterations, `total_head_loss` equals the total friction loss across the entire pipeline.

### Algorithm

```
1. 6 pipe sections: diameter, length, velocity lists

2. for i in range(6):
     D, L, V from lists[i]
     h_f = f*(L/D)*(V²/(2g))
     total_head_loss += h_f
     print row: section, D, V, h_f, cumulative

3. Print total head loss
```

### Expected output

```
Sec  D(mm)   V(m/s)   h_f(m)  Cumul.(m)
──────────────────────────────────────────
  1    300     1.20    0.117     0.117
  2    300     1.20    0.155     0.272
  3    250     1.80    0.534     0.806
...
Total head loss: X.XXX m
```

````python
import math

# Pipeline data: 6 sections
diameters_mm = [300, 300, 250, 250, 200, 200]
lengths_m    = [150, 200, 180, 120, 90,  100]
velocity_ms  = [1.2, 1.2, 1.8, 1.8, 2.8, 2.8]
f_darcy      = 0.018
g            = 9.81

total_head_loss = 0.0
print(f"{'Sec':>4} {'D(mm)':>6} {'V(m/s)':>8} {'h_f(m)':>8} {'Cumul.(m)':>10}")
print("─"*42)

for i in range(6):
    D   = diameters_mm[i] / 1000
    L   = lengths_m[i]
    V   = velocity_ms[i]
    h_f = f_darcy * (L/D) * (V**2 / (2*g))
    total_head_loss += h_f
    print(f"{i+1:>4} {diameters_mm[i]:>6} {V:>8.2f} {h_f:>8.3f} {total_head_loss:>10.3f}")

print(f"
Total head loss: {total_head_loss:.3f} m")
````

## Code Block 3 — Nested for Loops: SCS-CN Matrix

### What this code does

We compute an SCS-CN runoff table for 5 design storms × 4 curve numbers using nested for loops — an outer loop over storms and an inner loop over CN values.

### Why each step is taken

**Nested loops:**
The outer loop iterates over rows (storm depths). For each storm, the inner loop iterates over columns (CN values). This produces a 5×4 matrix of results — 20 Q values from 9 lines of code.

**`print(..., end="")`:**
Suppresses the automatic newline after each print, keeping values on the same row. `print()` with no arguments prints a newline — used at the end of each outer loop iteration.

### Algorithm

```
1. rainfalls_mm = [20,40,60,80,100]  (5 storms)
   cn_values    = [60,70,80,90]      (4 CNs)

2. Print header row

3. for CN in cn_values:
     S = 25400/CN - 254; Ia = 0.2*S

4. for P in rainfalls_mm:          (outer row loop)
     print P
     for CN in cn_values:          (inner column loop)
       compute Q with SCS-CN
       print Q (no newline)
     print() → newline at end of row
```

### Expected output

```
SCS-CN Runoff Depth (mm)
  CN:    60      70      80      90
P=  20   0.00    0.10    2.01   10.66
P=  40   0.36    4.94   13.57   27.17
...
```

````python
rainfalls_mm = [20, 40, 60, 80, 100]
cn_values    = [60, 70, 80, 90]

print("SCS-CN Runoff Depth (mm)")
print(f"{'':>6}", end="")
for CN in cn_values:
    print(f"  CN={CN:>2}", end="")
print()
print("─"*38)

for P in rainfalls_mm:
    print(f"P={P:>3}", end="")
    for CN in cn_values:
        S  = 25400/CN - 254
        Ia = 0.2 * S
        if P > Ia:
            Q = (P - Ia)**2 / (P - Ia + S)
        else:
            Q = 0.0
        print(f"  {Q:>6.2f}", end="")
    print()   # newline after each row
````

## Code Block 4 — for + if: IMD Daily Classification

### What this code does

We classify every day in June's 30-day rainfall record using for + if-elif-else and count the number of days in each IMD category.

### Why each step is taken

**Dictionary `counts`:**
Storing category counts in a dictionary avoids creating 6 separate counter variables. `counts["Heavy"] += 1` increments the count for the Heavy category.

**for + if-elif combined:**
The classification logic runs once per day — the for loop processes each day, and the if-elif-else classifies it. This is the most common pattern in data processing.

### Expected output

```
  Day  1:    0.0 mm → No rain
  Day  2:    0.0 mm → No rain
  Day  3:   12.4 mm → Light
  ...
Category counts: No rain=14  Light=4  Moderate=5  Heavy=4  Very Heavy=2  Ext.Heavy=1
```

````python
june_rainfall = [
    0.0, 0.0, 12.4, 45.6, 0.0, 8.2, 23.1,
    0.0, 0.0, 87.3, 34.5, 0.0, 56.2, 0.0,
    18.9, 0.0, 0.0, 134.5, 22.3, 45.6,
    0.0, 0.0, 67.8, 12.1, 0.0, 89.4, 33.2,
    0.0, 45.1, 28.7
]

counts = {"No rain":0,"Light":0,"Moderate":0,"Heavy":0,"Very Heavy":0,"Ext.Heavy":0}

for i, rain in enumerate(june_rainfall, 1):
    if rain == 0:           cat = "No rain"
    elif rain <= 15.5:      cat = "Light"
    elif rain <= 64.4:      cat = "Moderate"
    elif rain <= 115.5:     cat = "Heavy"
    elif rain <= 204.4:     cat = "Very Heavy"
    else:                   cat = "Ext.Heavy"
    counts[cat] += 1
    print(f"  Day {i:>2}: {rain:>6.1f} mm → {cat}")

print()
print("Category counts:", "  ".join(f"{k}={v}" for k,v in counts.items()))
````

## Session Summary — for Loops

| Pattern | Syntax | Use case |
|---|---|---|
| List iteration | `for item in list:` | Process each element |
| With index | `for i, item in enumerate(list,1):` | Need day/section number |
| Index-based | `for i in range(n):` | Access multiple parallel lists |
| Nested loops | `for a in ...: for b in ...:` | 2-D tables |
| No newline | `print(..., end="")` | Same-row printing |
| Accumulate | `total += value` | Running sum inside loop |
| Count per category | `counts[cat] += 1` | Frequency table |

July 2024 rainfall — compute total, heavy days, peak, and dry day percentage.

### ▶ Assignment cell

````python
july = [45.2, 0, 123.4, 67.8, 0, 0, 89.1, 34.5, 12.3, 0,
       156.7, 45.2, 0, 0, 78.9, 34.1, 0, 89.2, 45.6, 0,
       0, 67.3, 12.4, 0, 89.5, 34.2, 178.9, 0, 45.1, 23.4, 0]

total = 0; heavy_days = 0; peak = 0; peak_day = 0; dry = 0

for i, rain in enumerate(july, 1):
    total += rain
    if rain == 0:
        dry += 1
    if rain >= 64.5:
        heavy_days += 1
    if rain > peak:
        peak = rain; peak_day = i

pct_dry = dry / len(july) * 100
print(f"Monthly total  : {total:.1f} mm")
print(f"Heavy+ days    : {heavy_days}")
print(f"Peak           : {peak} mm on Day {peak_day}")
print(f"Dry days       : {dry} ({pct_dry:.1f}%)")
````
