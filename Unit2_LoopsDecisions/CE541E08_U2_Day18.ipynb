### ▶ Part A — Rainfall Data Analysis (8 marks)

```python
# ═══════════════════════════════════════════════════════
# CIA I — PART A: Rainfall Analysis
# Analyse the monsoon rainfall record below using loops
# and conditional statements.
# ═══════════════════════════════════════════════════════

# Bengaluru monsoon rainfall (Jun-Sep 2023), mm/day
# 122 days of data
monsoon_rainfall = [
    # June (30 days)
    0,0,8.2,34.5,0,12.4,45.6,0,0,67.8,23.4,0,89.2,12.3,
    0,134.5,45.6,0,0,78.9,34.2,0,156.7,89.3,12.4,0,45.6,23.4,
    0,67.8,
    # July (31 days)
    89.2,156.7,0,234.5,178.9,45.6,12.3,0,89.4,267.8,134.5,0,
    45.6,178.3,89.2,0,0,56.7,189.4,234.5,89.3,0,45.6,0,178.9,
    89.2,34.5,0,156.7,89.4,45.6,
    # August (31 days)
    123.4,0,234.5,89.2,156.7,0,45.6,178.9,89.3,0,234.8,156.7,
    0,89.2,45.6,0,178.9,89.4,0,0,123.5,67.8,0,45.6,178.3,89.2,
    0,234.5,89.3,45.6,0,
    # September (30 days)
    78.9,0,45.6,123.4,0,0,89.2,34.5,0,156.7,89.3,0,45.6,23.4,
    0,78.9,34.2,0,89.5,45.6,12.3,0,67.8,34.5,0,89.2,45.6,0,
    23.4,12.3
]

months     = ["June"]*30 + ["July"]*31 + ["August"]*31 + ["September"]*30
month_days = {"June":30,"July":31,"August":31,"September":30}

# -- TASK A1: Overall statistics --------------------------
total_monsoon  = sum(monsoon_rainfall)
mean_daily     = total_monsoon / len(monsoon_rainfall)
peak_val       = max(monsoon_rainfall)
peak_idx       = monsoon_rainfall.index(peak_val)
peak_month     = months[peak_idx]
dry_days       = monsoon_rainfall.count(0)
heavy_days     = sum(1 for r in monsoon_rainfall if r >= 64.5)

print("PART A1 — MONSOON SEASON OVERVIEW")
print("-"*50)
print(f"Total monsoon rainfall : {total_monsoon:.1f} mm")
print(f"Mean daily             : {mean_daily:.2f} mm/day")
print(f"Peak day               : {peak_val} mm ({peak_month}, Day {peak_idx+1})")
print(f"Dry days               : {dry_days} of {len(monsoon_rainfall)}")
print(f"Heavy+ rain days       : {heavy_days}")

# -- TASK A2: Monthly breakdown ---------------------------
print("PART A2 — MONTHLY BREAKDOWN")
print("-"*50)
start = 0
for month, days in month_days.items():
    month_data = monsoon_rainfall[start:start+days]
    m_total    = sum(month_data)
    m_peak     = max(month_data)
    m_heavy    = sum(1 for r in month_data if r >= 64.5)
    print(f"{month:<12}: {m_total:>7.1f} mm  peak:{m_peak:>7.1f} mm  heavy:{m_heavy:>3} days")
    start += days

# --- INSTRUCTOR NOTE ---------------------------------------
# ANSWER KEY:
# Total ≈ 3847 mm, mean ≈ 31.5 mm/day
# Peak: August (exact depends on position in data)
# -----------------------------------------------------------
```

### ▶ Part B — IMD Classification Summary (4 marks)

```python
monsoon_rainfall = [
    0,0,8.2,34.5,0,12.4,45.6,0,0,67.8,23.4,0,89.2,12.3,
    0,134.5,45.6,0,0,78.9,34.2,0,156.7,89.3,12.4,0,45.6,23.4,
    0,67.8,89.2,156.7,0,234.5,178.9,45.6,12.3,0,89.4,267.8,134.5,0,
    45.6,178.3,89.2,0,0,56.7,189.4,234.5,89.3,0,45.6,0,178.9,
    89.2,34.5,0,156.7,89.4,45.6,123.4,0,234.5,89.2,156.7,0,45.6,178.9,89.3,0,234.8,156.7,
    0,89.2,45.6,0,178.9,89.4,0,0,123.5,67.8,0,45.6,178.3,89.2,
    0,234.5,89.3,45.6,0,78.9,0,45.6,123.4,0,0,89.2,34.5,0,156.7,89.3,0,45.6,23.4,
    0,78.9,34.2,0,89.5,45.6,12.3,0,67.8,34.5,0,89.2,45.6,0,23.4,12.3
]

# Classify all days and count each category
counts = {"No rain":0,"Light":0,"Moderate":0,"Heavy":0,"Very Heavy":0,"Extremely Heavy":0}

for rain in monsoon_rainfall:
    if rain == 0:             counts["No rain"] += 1
    elif rain <= 15.5:        counts["Light"] += 1
    elif rain <= 64.4:        counts["Moderate"] += 1
    elif rain <= 115.5:       counts["Heavy"] += 1
    elif rain <= 204.4:       counts["Very Heavy"] += 1
    else:                     counts["Extremely Heavy"] += 1

print("PART B — IMD CATEGORY DISTRIBUTION")
print("-"*50)
total_days = len(monsoon_rainfall)
for cat, ct in counts.items():
    bar  = "█" * ct
    pct  = ct/total_days*100
    print(f"  {cat:<22}: {ct:>3} days ({pct:>5.1f}%)  {bar}")
print("-"*50)
print(f"  {'TOTAL':<22}: {total_days} days")
```

### ▶ Part C — Storm drain design using Rational Method (4 marks)

```python
# ═══════════════════════════════════════════════════════
# PART C: Design a stormwater drain for a new Bengaluru
# residential layout using the Rational Method.
# ═══════════════════════════════════════════════════════

import math

# Layout data
area_ha      = 18.5     # total layout area, ha
c_roads      = 0.90     # roads and pavements (35% of area)
c_roofs      = 0.85     # rooftops (30% of area)
c_gardens    = 0.35     # green areas (35% of area)
area_split   = [0.35, 0.30, 0.35]   # fractions

# Design storm: 25-year return period (Bengaluru)
intensity_mmhr = 62.0   # mm/hr (standard design value)

# Drain hydraulics
n_concrete   = 0.013
slope        = 0.005

# Weighted runoff coefficient
C_weighted = (c_roads * area_split[0] +
              c_roofs * area_split[1] +
              c_gardens * area_split[2])

# Rational method
total_area_m2 = area_ha * 10000
intensity_ms  = intensity_mmhr / 1000 / 3600
Q_peak        = C_weighted * intensity_ms * total_area_m2

# Size a circular drain using Manning's equation
# Try diameters until one provides adequate capacity
print("PART C — STORMWATER DRAIN DESIGN")
print("-"*55)
print(f"Layout area    : {area_ha} ha")
print(f"Weighted C     : {C_weighted:.3f}")
print(f"Design intensity: {intensity_mmhr} mm/hr (25-yr RP)")
print(f"Peak discharge : {Q_peak:.4f} m³/s  ({Q_peak*1000:.2f} L/s)")
print()
print("Pipe sizing (Manning's equation, n=0.013, S=1:200):")
print(f"{'D(mm)':>8} {'Q_cap(L/s)':>12} {'V(m/s)':>10} {'Adequate?'}")
print("-"*40)

for D_mm in [300, 375, 450, 525, 600]:
    D     = D_mm/1000
    A     = math.pi*(D/2)**2
    P     = math.pi*D
    R     = A/P
    V     = (1/n_concrete) * R**(2/3) * slope**0.5
    Q_cap = V*A
    ok    = "YES ✓" if Q_cap >= Q_peak else "no"
    print(f"{D_mm:>8} {Q_cap*1000:>12.2f} {V:>10.3f} {ok}")
```
