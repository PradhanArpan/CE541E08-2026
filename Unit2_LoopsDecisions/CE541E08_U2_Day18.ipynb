## Part A — Rainfall Analysis: Monthly Breakdown

### What this code does

We process 122 days of monsoon data (June 30 + July 31 + August 31 + September 30), split it into monthly sub-arrays, and compute total, mean, peak, and heavy-day count for each month.

### Why each step is taken

**Month boundary indices:**
June = days 0–29 (30 days), July = 30–60 (31 days), August = 61–91 (31 days), September = 92–121 (30 days). List slicing extracts each month.

**`for month_name, month_data in zip(months, [june,...]):`:**
`zip` pairs each month name with its data sub-array. Inside the loop, all statistics are computed for one month at a time.

**Heavy day count inside the loop:**
`sum(1 for r in month_data if r >= 64.5)` uses a generator expression — a compact way to count elements meeting a condition.

### Expected output

```
==================================================
PART A — MONTHLY RAINFALL BREAKDOWN
--------------------------------------------------
Month    Total(mm)  Mean(mm)  Peak(mm)  Heavy+days
--------------------------------------------------
June       ...        ...       ...        ...
July       ...        ...       ...        ...
August     ...        ...       ...        ...
September  ...        ...       ...        ...
SEASON     ...        ...       ...        ...
```

````python
monsoon_rainfall = [
    # June (30 days)
    0,0,8.2,34.5,0,12.4,45.6,0,0,67.8,23.4,0,89.2,12.3,
    0,134.5,45.6,0,0,78.9,34.2,0,156.7,89.3,12.4,0,45.6,23.4,
    0,67.8,
    # July (31 days)
    89.2,156.7,0,234.5,178.9,45.6,12.3,0,89.4,267.8,134.5,0,
    45.6,178.3,89.2,0,0,56.7,189.4,234.5,89.3,0,45.6,0,178.9,
    89.2,34.5,0,156.7,89.4,45.6,
    # August (31 days)
    123.4,0,234.5,89.2,156.7,0,45.6,178.9,89.3,0,234.8,156.7,
    0,89.2,45.6,0,178.9,89.4,0,0,123.5,67.8,0,45.6,178.3,89.2,
    0,234.5,89.3,45.6,0,
    # September (30 days)
    78.9,0,45.6,123.4,0,0,89.2,34.5,0,156.7,89.3,0,45.6,23.4,
    0,67.8,0,45.6,89.2,0,234.5,89.3,45.6,0,78.9,0,45.6,123.4,
    0,89.2
]

months     = ["June","July","August","September"]
boundaries = [(0,30),(30,61),(61,92),(92,122)]

print("="*50)
print("PART A — MONTHLY RAINFALL BREAKDOWN")
print("-"*50)
print(f"{'Month':<12} {'Total':>9} {'Mean':>8} {'Peak':>8} {'Heavy+':>10}")
print("-"*50)

season_total = 0
for month_name, (start,end) in zip(months, boundaries):
    data      = monsoon_rainfall[start:end]
    total     = sum(data)
    mean      = total / len(data)
    peak      = max(data)
    heavy     = sum(1 for r in data if r >= 64.5)
    season_total += total
    print(f"{month_name:<12} {total:>9.1f} {mean:>8.1f} {peak:>8.1f} {heavy:>10}")
print("-"*50)
season_mean = season_total / len(monsoon_rainfall)
season_peak = max(monsoon_rainfall)
season_h    = sum(1 for r in monsoon_rainfall if r >= 64.5)
print(f"{'SEASON':<12} {season_total:>9.1f} {season_mean:>8.1f} {season_peak:>8.1f} {season_h:>10}")
````

## Part B — IMD Category Analysis: 122-Day Record

### What this code does

We classify all 122 days into IMD categories and print a frequency and percentage table.

### Expected output

```
==================================================
PART B — IMD CATEGORY DISTRIBUTION (122 days)
--------------------------------------------------
Category          Days    %   Total_mm
--------------------------------------------------
No rain             XX  XX.X%     0.0
Light               XX  XX.X%    XX.X
...
TOTAL              122 100.0%   XXXX.X
```

````python
# IMD Classification of all 122 days
categories_list = ["No rain","Light","Moderate","Heavy","Very Heavy","Ext. Heavy"]
counts   = {c:0   for c in categories_list}
totals_r = {c:0.0 for c in categories_list}

for rain in monsoon_rainfall:
    if rain == 0:         cat = "No rain"
    elif rain <= 15.5:    cat = "Light"
    elif rain <= 64.4:    cat = "Moderate"
    elif rain <= 115.5:   cat = "Heavy"
    elif rain <= 204.4:   cat = "Very Heavy"
    else:                 cat = "Ext. Heavy"
    counts[cat]   += 1
    totals_r[cat] += rain

n = len(monsoon_rainfall)
print("="*50)
print("PART B — IMD CATEGORY DISTRIBUTION (122 days)")
print("-"*50)
print(f"{'Category':<16} {'Days':>5} {'%':>6}   {'Total_mm':>10}")
print("-"*50)
for c in categories_list:
    print(f"{c:<16} {counts[c]:>5} {counts[c]/n*100:>5.1f}%   {totals_r[c]:>10.1f}")
print("-"*50)
print(f"{'TOTAL':<16} {n:>5} {'100.0%':>6}   {sum(totals_r.values()):>10.1f}")
````

## Part C — Stormwater Drain Design

### What this code does

We design a stormwater drain for an 18.5 ha residential layout using the Rational Method and check the capacity using Manning's formula.

### Expected output

```
==================================================
PART C — STORMWATER DRAIN DESIGN
--------------------------------------------------
Composite C      : 0.XX
Design discharge : XX.XX L/s
Manning's discharge: XX.XX L/s
Design adequate  : True/False
Recommended D    : XXX mm
```

````python
import math

area_ha      = 18.5
c_roads      = 0.90; area_split_roads   = 0.35
c_roofs      = 0.85; area_split_roofs   = 0.30
c_gardens    = 0.35; area_split_gardens = 0.35
intensity_mmhr = 52.0   # design rainfall intensity (IMD Heavy Rain approx)

# Composite C
C_composite = (c_roads*area_split_roads + c_roofs*area_split_roofs +
               c_gardens*area_split_gardens)

# Peak discharge (Rational Method)
area_m2      = area_ha * 10000
intensity_ms = intensity_mmhr / 1000 / 3600
Q_peak       = C_composite * intensity_ms * area_m2

print("="*50)
print("PART C — STORMWATER DRAIN DESIGN")
print("-"*50)
print(f"Composite C      : {C_composite:.3f}")
print(f"Design discharge : {Q_peak*1000:.2f} L/s")

# Manning's capacity check for candidate diameters
n_rough = 0.013; S = 0.001
print(f"
{'D(mm)':>6} {'V(m/s)':>8} {'Q(L/s)':>8} {'Adequate'}")
for D_mm in [300, 350, 400, 450, 500]:
    D = D_mm/1000; A = math.pi*(D/2)**2; R = D/4
    V = (1/n_rough)*R**(2/3)*S**0.5; Q = V*A
    ok = Q >= Q_peak
    print(f"{D_mm:>6} {V:>8.3f} {Q*1000:>8.2f} {str(ok):>10}")
````
