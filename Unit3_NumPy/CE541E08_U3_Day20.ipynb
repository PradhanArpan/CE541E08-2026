## Section 1 — Array Dimensions

Yesterday we worked with 1-D arrays — a single row of values like 30 daily rainfall readings.
Real engineering data is often multi-dimensional:

| Dimension | Shape example | Engineering example |
|---|---|---|
| **1-D** | `(12,)` | Monthly totals for one year |
| **2-D** | `(3, 12)` | Monthly totals for 3 years — rows = years, cols = months |
| **3-D** | `(3, 4, 5)` | 3 years × 4 seasons × 5 sub-basins |

The key NumPy concept introduced today is the **axis** parameter:
- `axis=0` means operate **across rows** (down each column)
- `axis=1` means operate **across columns** (along each row)

For a rainfall matrix where rows = years and cols = months:
- `data.sum(axis=1)` → sums across the 12 months → gives one annual total per year
- `data.mean(axis=0)` → averages across the years → gives one long-term mean per month

## Code Block 1 — 1-D and 2-D Arrays

### What this code does

We create a 1-D array of monthly rainfall totals for one year, then extend it to a 2-D array covering 3 years. We use the `axis` parameter to compute annual totals (one per year) and monthly climatology (one per month).

### Why each step is taken

**1-D array — `monthly_total`:**
A straightforward list of 12 monthly values. Shape `(12,)` means one row, 12 columns. Good for a single year. Limited for multi-year analysis.

**2-D array — `annual_data`:**
Rows represent years (2022, 2023, 2024). Columns represent months (Jan to Dec). Shape `(3, 12)` — 3 rows, 12 columns. This structure lets NumPy compute across years or across months with a single call.

**`annual_data.sum(axis=1)` — annual totals:**
`axis=1` means "sum across the column direction" — i.e. add all 12 months together for each row. The result has one value per row = one annual total per year. Think of it as collapsing the 12 columns into one.

**`annual_data.mean(axis=0)` — monthly climatology:**
`axis=0` means "average across the row direction" — i.e. average the 3 years together for each column. The result has one value per column = one long-term mean per month. This is how a 30-year climatology is computed in practice.

### Algorithm

```
1. Create 1-D array: monthly_total (12 values)
   Print its shape and total

2. Create 2-D array: annual_data — shape (3, 12)
   rows = years 2022, 2023, 2024
   cols = months Jan to Dec

3. annual_data.sum(axis=1)
   → sums across columns (months) for each row (year)
   → result shape: (3,) — one annual total per year

4. annual_data.mean(axis=0)
   → averages across rows (years) for each column (month)
   → result shape: (12,) — one mean per month
   → this is the monthly climatology

5. Print all results
```

### Expected output

```
1-D shape: (12,)   total: 765
2-D shape: (3, 12)
Annual totals (axis=1): [765 756 783]
Monthly means (axis=0): [  7.7  11.7  18.7  51.7  89.  135.7 118.3 113.3  94.  71.   43.3  13. ]
```

````python
import numpy as np

# 1-D array: 12 monthly rainfall totals (mm) for one year
# Shape (12,) — a single row of 12 values
monthly_total = np.array([8, 12, 18, 52, 87, 134, 118, 113, 95, 71, 44, 13])
print("1-D shape:", monthly_total.shape, "  total:", monthly_total.sum())

# 2-D array: 3 years × 12 months — each row is one year
# Shape (3, 12) — 3 rows, 12 columns
annual_data = np.array([
    [8,  12, 18, 52, 87, 134, 118, 113, 95, 71, 44, 13],   # 2022
    [5,   8, 22, 48, 92, 145, 123, 108, 89, 68, 38, 10],   # 2023
    [10, 15, 16, 55, 83, 128, 115, 119, 98, 74, 48, 16],   # 2024
])
print("2-D shape:", annual_data.shape)

# axis=1: sum ACROSS columns (across the 12 months) for each row (year)
# Result: one value per row = annual total for each year
print("Annual totals (axis=1):", annual_data.sum(axis=1))

# axis=0: average ACROSS rows (across the 3 years) for each column (month)
# Result: one value per column = long-term monthly mean
print("Monthly means (axis=0):", annual_data.mean(axis=0).round(1))
````

## Code Block 2 — 2-D Grid: Spatial Rainfall over a Catchment

### What this code does

We represent daily rainfall across a 5×5 grid of sub-basins as a 2-D array. Rows represent the north-south direction, columns represent east-west. We compute spatial statistics and use boolean masking to identify which grid cells received Heavy Rain.

### Why each step is taken

**5×5 grid structure:**
In practice, rainfall data from satellite or radar products comes as a spatial grid — each cell represents a geographic area. A 2-D NumPy array maps directly to this structure. Row index = latitude band, column index = longitude band.

**`axis=1` for row totals:**
Summing across columns (axis=1) gives the total rainfall in each north-south strip. This tells you which latitudinal band received the most rain — useful for identifying where a storm made landfall.

**`axis=0` for column totals:**
Summing across rows (axis=0) gives the total rainfall in each east-west strip — identifying the longitudinal spread of the storm.

**`rainfall_grid >= 64.5` — spatial boolean mask:**
Every cell in the grid is compared to the IMD Heavy Rain threshold simultaneously. The result is a 5×5 True/False grid. `.sum()` on a 2-D boolean array counts all True values across the entire grid — giving the total number of Heavy Rain cells regardless of their location.

### Algorithm

```
1. Create 5×5 rainfall grid — shape (5, 5)
   rows = north to south, cols = west to east

2. rainfall_grid.mean()   → spatial mean over all 25 cells
   rainfall_grid.sum(axis=1) → row totals (N→S strips)
   rainfall_grid.sum(axis=0) → col totals (W→E strips)

3. Boolean mask: rainfall_grid >= 64.5
   → produces a 5×5 True/False grid
   → heavy.sum() counts all True cells across the grid

4. Print all spatial statistics
```

### Expected output

```
Shape       : (5, 5)  (5x5 sub-basins)
Spatial mean: 66.08 mm
Row totals  : [336.8 362.3 449.3 348.  168.9]
Col totals  : [124.1 272.4 504.3 426.4 303. ]
Heavy cells : 9 of 25
```

````python
import numpy as np

# 5×5 grid of daily rainfall (mm) over Cauvery catchment sub-basins
# Each row = a north-south strip; each column = an east-west strip
rainfall_grid = np.array([
    [45.2, 67.8, 89.3, 78.4, 56.1],
    [34.5, 78.9, 123.4, 89.2, 67.3],
    [23.1, 56.7, 145.6, 134.5, 89.4],
    [12.4, 45.6, 89.3, 78.5, 56.2],
    [8.9, 23.4, 56.7, 45.8, 34.1],
])

print(f"Shape       : {rainfall_grid.shape}  (5x5 sub-basins)")

# .mean() with no axis = mean of ALL 25 values (the entire grid)
print(f"Spatial mean: {rainfall_grid.mean():.2f} mm")

# axis=1: sum across columns for each row → total per N-S strip
print(f"Row totals  : {rainfall_grid.sum(axis=1).round(1)}")

# axis=0: sum across rows for each column → total per E-W strip
print(f"Col totals  : {rainfall_grid.sum(axis=0).round(1)}")

# Boolean mask: compare every cell to the IMD Heavy Rain threshold
# heavy is a 5×5 True/False grid — True where rainfall >= 64.5 mm
heavy = rainfall_grid >= 64.5
# .sum() on a boolean grid counts all True cells
print(f"Heavy cells : {heavy.sum()} of {rainfall_grid.size}")
````

## Code Block 3 — 3-D Array: Years × Seasons × Basins

### What this code does

We extend the concept to three dimensions — 3 years × 4 seasons × 5 sub-basins. This is the structure used in climate datasets where you have time, season, and space all in one array. We extract the monsoon season slice and compute the 3-year mean for each basin.

### Why each step is taken

**3-D array shape `(3, 4, 5)`:**
- Axis 0 (size 3): years — 2022, 2023, 2024
- Axis 1 (size 4): seasons — pre-monsoon, monsoon, post-monsoon, winter
- Axis 2 (size 5): sub-basins — Hemavathi, Harangi, Kabini, Suvarnavathi, KRS

**`base[:,1,:]` — selecting the monsoon season:**
The `:` means "all values along this axis". So `base[:,1,:]` means: all years (axis 0), season index 1 = monsoon (axis 1), all basins (axis 2). The result is a `(3, 5)` array — 3 years × 5 basins, monsoon season only.

**`.mean(axis=0)`:**
Averaging across axis 0 (years) gives the 3-year average monsoon rainfall for each of the 5 basins. The result is a `(5,)` array — one mean per basin.

### Algorithm

```
1. Create 3-D array — shape (3, 4, 5)
   axis 0 = year (3 values)
   axis 1 = season (4 values: pre-monsoon, monsoon, post-monsoon, winter)
   axis 2 = sub-basin (5 values)

2. base[:,1,:] — slice out the monsoon season
   : in axis 0 = all 3 years
   1 in axis 1 = season index 1 (monsoon)
   : in axis 2 = all 5 basins
   Result shape: (3, 5)

3. .mean(axis=0) — average across the 3 years
   Result shape: (5,) — one mean per basin

4. Print mean monsoon rainfall for each named basin
```

### Expected output

```
Shape: (3, 4, 5)  (3 years x 4 seasons x 5 basins)
Mean monsoon rainfall by sub-basin (3-year avg):
  Hemavathi      : 45.7 mm
  Harangi        : 52.7 mm
  Kabini         : 46.0 mm
  Suvarnavathi   : 55.0 mm
  KRS            : 50.3 mm
```

````python
import numpy as np

# 3-D array: 3 years × 4 seasons × 5 sub-basins
# Axis 0 = years (2022, 2023, 2024)
# Axis 1 = seasons (pre-monsoon, monsoon, post-monsoon, winter)
# Axis 2 = sub-basins (Hemavathi, Harangi, Kabini, Suvarnavathi, KRS)
base = np.array([
    [[15,12,18,10,14], [45,52,48,55,50], [85,92,78,95,88], [30,28,35,25,32]],  # 2022
    [[12,10,16, 8,11], [50,58,52,60,54], [90,95,82,100,92],[28,25,32,22,30]],  # 2023
    [[18,15,20,12,16], [42,48,45,52,47], [80,88,75,90,85], [33,30,38,28,35]],  # 2024
])
print(f"Shape: {base.shape}  (3 years x 4 seasons x 5 basins)")

# base[:,1,:] — extract monsoon season (index 1) across all years and basins
# : in position 0 = all 3 years
# 1 in position 1 = season index 1 (monsoon)
# : in position 2 = all 5 basins
# Result shape: (3, 5)
monsoon_data = base[:, 1, :]

# .mean(axis=0): average across the year axis → one value per basin
# Result shape: (5,) — the 3-year mean monsoon rainfall per basin
monsoon_mean = monsoon_data.mean(axis=0)

basins = ['Hemavathi', 'Harangi', 'Kabini', 'Suvarnavathi', 'KRS']
print("Mean monsoon rainfall by sub-basin (3-year avg):")
for b, v in zip(basins, monsoon_mean):
    print(f"  {b:<15}: {v:.1f} mm")
````

## Code Block 4 — Matrix Operations: Solving a Structural System

### What this code does

We set up a simplified 3-node structural stiffness matrix K and a load vector F, then solve the system K × u = F for the nodal displacements u using NumPy's linear algebra solver. This is the same mathematical operation performed by structural analysis software (STAAD, SAP2000) for every load case.

### Why each step is taken

**Stiffness matrix K:**
In the stiffness method, K is a square matrix where each entry represents the force required at one node due to unit displacement at another. For a 3-node system, K is 3×3. The matrix is assembled from individual element stiffness values.

**Load vector F:**
A 1-D array of forces applied at each node. Here, 50 kN is applied at node 2 (index 1); nodes 1 and 3 have no external load.

**`np.linalg.solve(K, F)`:**
Solves K × u = F for u. This is more numerically stable than computing the inverse K⁻¹ directly (which is what a student might try with `np.linalg.inv(K) @ F`). NumPy uses LU decomposition internally.

**`K @ u` — verification:**
The `@` operator performs matrix multiplication. If the solution is correct, K × u should equal F. We round to 4 decimal places to handle floating-point precision.

### Algorithm

```
1. Define stiffness matrix K — shape (3, 3)
   Values represent spring stiffness in kN/m

2. Define load vector F — shape (3,)
   F = [0, 50, 0] kN (force at node 2 only)

3. u = np.linalg.solve(K, F)
   Solves K × u = F using LU decomposition
   u = displacement at each node (metres)

4. Verify: K @ u should equal F
   @ is the matrix multiplication operator in NumPy

5. Find max displacement and which node it occurs at
   abs(u).max()    → largest displacement magnitude
   abs(u).argmax() → index of that node (0-based) + 1 = node number
```

### Expected output

```
Stiffness K:
[[ 400. -200.    0.]
 [-200.  500. -300.]
 [   0. -300.  300.]]
Displacements u (m): [0.25   0.5    0.5  ]
Verify K*u = [ 0. 50.  0.]
Max disp: 500.00 mm at node 2
```

````python
import numpy as np

# 3×3 stiffness matrix K (kN/m)
# Each off-diagonal entry is negative (coupling between adjacent nodes)
K = np.array([
    [ 400, -200,    0],
    [-200,  500, -300],
    [   0, -300,  300],
], dtype=float)   # dtype=float ensures decimal arithmetic

# Load vector F (kN) — 50 kN applied at node 2 (index 1)
F = np.array([0, 50, 0], dtype=float)

# Solve K × u = F for displacements u (metres)
# np.linalg.solve is numerically more stable than computing inv(K) @ F
u = np.linalg.solve(K, F)

print("Stiffness K:"); print(K)
print(f"Displacements u (m): {np.round(u, 4)}")

# Verification: K @ u should equal F
# @ is the matrix multiplication operator in Python/NumPy
print(f"Verify K*u = {np.round(K @ u, 4)}")

# Maximum displacement — abs() handles negative displacements correctly
print(f"Max disp: {abs(u).max()*1000:.2f} mm at node {abs(u).argmax()+1}")
````

## Session Summary — New Concepts Introduced Today

| Concept | Syntax | What it gives |
|---|---|---|
| 1-D array | `np.array([v1, v2, ..., v12])` | Shape `(12,)` — one row |
| 2-D array | `np.array([[row1], [row2], ...])` | Shape `(r, c)` — r rows, c cols |
| 3-D array | Nested lists 3 levels deep | Shape `(a, b, c)` |
| Sum across rows (per year) | `data.sum(axis=1)` | One total per row |
| Mean across years (per month) | `data.mean(axis=0)` | One mean per column |
| Slice a 3-D array | `base[:, 1, :]` | All years, season 1, all basins |
| Matrix solve | `np.linalg.solve(K, F)` | Displacements u from K × u = F |
| Matrix multiply | `K @ u` | Matrix product |
| 2-D boolean mask | `grid >= 64.5` | True/False at every cell |
| Count True in 2-D | `mask.sum()` | Total number of True cells |
