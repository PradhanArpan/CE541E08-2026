## Section 1 — Array dimensions

| Dimension | Shape | Engineering use |
|---|---|---|
| 1-D | `(30,)` | Daily rainfall, one month |
| 2-D | `(12, 30)` | Monthly rainfall matrix |
| 3-D | `(10, 12, 30)` | 10-year monthly daily rainfall |

### ▶ Cell 1 — 1-D and 2-D arrays

````python
import numpy as np
monthly_total = np.array([8,12,18,52,87,134,118,113,95,71,44,13])
print("1-D shape:", monthly_total.shape, "  total:", monthly_total.sum())
annual_data = np.array([
    [8,12,18,52,87,134,118,113,95,71,44,13],
    [5,8,22,48,92,145,123,108,89,68,38,10],
    [10,15,16,55,83,128,115,119,98,74,48,16],
])
print("2-D shape:", annual_data.shape)
print("Annual totals (axis=1):", annual_data.sum(axis=1))
print("Monthly means (axis=0):", annual_data.mean(axis=0).round(1))
# --- INSTRUCTOR NOTE ---
# axis=1 sums ACROSS columns (months) -> per year
# axis=0 means ACROSS rows (years) -> monthly climatology
````

### ▶ Cell 2 — 2-D: rainfall grid over catchment

````python
import numpy as np
rainfall_grid = np.array([
    [45.2,67.8,89.3,78.4,56.1],
    [34.5,78.9,123.4,89.2,67.3],
    [23.1,56.7,145.6,134.5,89.4],
    [12.4,45.6,89.3,78.5,56.2],
    [8.9,23.4,56.7,45.8,34.1],
])
print(f"Shape       : {rainfall_grid.shape}  (5x5 sub-basins)")
print(f"Spatial mean: {rainfall_grid.mean():.2f} mm")
print(f"Row totals  : {rainfall_grid.sum(axis=1).round(1)}")
print(f"Col totals  : {rainfall_grid.sum(axis=0).round(1)}")
heavy = rainfall_grid >= 64.5
print(f"Heavy cells : {heavy.sum()} of {rainfall_grid.size}")
````

### ▶ Cell 3 — 3-D array: years x seasons x basins

````python
import numpy as np
base = np.array([
    [[15,12,18,10,14],[45,52,48,55,50],[85,92,78,95,88],[30,28,35,25,32]],
    [[12,10,16,8,11],[50,58,52,60,54],[90,95,82,100,92],[28,25,32,22,30]],
    [[18,15,20,12,16],[42,48,45,52,47],[80,88,75,90,85],[33,30,38,28,35]],
])
print(f"Shape: {base.shape}  (3 years x 4 seasons x 5 basins)")
basins = ['Hemavathi','Harangi','Kabini','Suvarnavathi','KRS']
monsoon_mean = base[:,1,:].mean(axis=0)
for b,v in zip(basins,monsoon_mean):
    print(f"  {b:<15}: {v:.1f} mm")
# --- INSTRUCTOR NOTE ---
# base[:,1,:] selects ALL years, season index 1 (monsoon), ALL basins
````

### ▶ Cell 4 — Matrix solve: structural analysis

````python
import numpy as np
K = np.array([[400,-200,0],[-200,500,-300],[0,-300,300]], dtype=float)
F = np.array([0,50,0], dtype=float)
u = np.linalg.solve(K,F)
print("Stiffness K:"); print(K)
print(f"Displacements u (m): {np.round(u,4)}")
print(f"Verify K*u = {np.round(K@u,4)}")
print(f"Max disp: {abs(u).max()*1000:.2f} mm at node {abs(u).argmax()+1}")
````
